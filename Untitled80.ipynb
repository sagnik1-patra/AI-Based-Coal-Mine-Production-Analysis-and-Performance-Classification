```python
import os
import re
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout
)

from tensorflow.keras.callbacks import EarlyStopping


# =============================================================================
# 2. SETTINGS
# =============================================================================

warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# =============================================================================
# 3. PROJECT PATHS
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Coal Mine Production Analysis and Performance Classification"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "rs_session-241_au482_1.1.csv"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 4. OUTPUT PATHS
# =============================================================================

ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "accuracy_graph.png"
)

PREDICTION_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "prediction.csv"
)

PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "prediction_graph.png"
)

RESULT_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "result.csv"
)

RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "result_graph.png"
)

COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "comparison_graph.png"
)

HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "heatmap.png"
)

PROCESSED_DATA_PATH = os.path.join(
    PROJECT_DIR,
    "processed_coal_production_data.csv"
)


# =============================================================================
# 5. CHECK DATASET
# =============================================================================

print("=" * 80)
print("AI-BASED COAL MINE PRODUCTION ANALYSIS")
print("=" * 80)


if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}"
    )


print("\nDataset found:")
print(CSV_PATH)


# =============================================================================
# 6. LOAD DATASET
# =============================================================================

print("\n" + "=" * 80)
print("LOADING DATASET")
print("=" * 80)


encodings = [
    "utf-8",
    "utf-8-sig",
    "cp1252",
    "latin1"
]


df = None
used_encoding = None


for encoding in encodings:

    try:

        df = pd.read_csv(
            CSV_PATH,
            encoding=encoding
        )

        used_encoding = encoding

        break

    except UnicodeDecodeError:

        continue


if df is None:

    raise ValueError(
        "Could not read the CSV file."
    )


print("\nDataset loaded successfully.")

print("\nEncoding:")
print(used_encoding)

print("\nDataset shape:")
print(df.shape)


# =============================================================================
# 7. CLEAN COLUMN NAMES
# =============================================================================

def clean_column_name(column):

    column = str(column)

    column = column.replace(
        "\n",
        " "
    )

    column = column.replace(
        "\r",
        " "
    )

    column = re.sub(
        r"\s+",
        " ",
        column
    )

    return column.strip()


df.columns = [
    clean_column_name(col)
    for col in df.columns
]


# =============================================================================
# 8. MAKE COLUMN NAMES UNIQUE
# =============================================================================

def make_unique_columns(columns):

    counts = {}

    new_columns = []

    for column in columns:

        if column not in counts:

            counts[column] = 0

            new_columns.append(
                column
            )

        else:

            counts[column] += 1

            new_name = (
                f"{column}_{counts[column]}"
            )

            while new_name in new_columns:

                counts[column] += 1

                new_name = (
                    f"{column}_{counts[column]}"
                )

            new_columns.append(
                new_name
            )

    return new_columns


df.columns = make_unique_columns(
    df.columns
)


print("\nColumns:")

for i, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{i}. {column}"
    )


# =============================================================================
# 9. REMOVE COMPLETELY EMPTY DATA
# =============================================================================

df = df.dropna(
    axis=0,
    how="all"
)

df = df.dropna(
    axis=1,
    how="all"
)


# =============================================================================
# 10. COLUMN SEARCH FUNCTION
# =============================================================================

def find_column(
    columns,
    keywords,
    exclude=None
):

    if exclude is None:

        exclude = []


    # Exact matches first

    for keyword in keywords:

        for column in columns:

            if column in exclude:
                continue

            if (
                str(column).lower().strip()
                ==
                keyword.lower().strip()
            ):

                return column


    # Partial matches

    for keyword in keywords:

        for column in columns:

            if column in exclude:
                continue

            if (
                keyword.lower()
                in str(column).lower()
            ):

                return column


    return None


# =============================================================================
# 11. DETECT PRODUCTION COLUMN
# =============================================================================

production_column = find_column(

    df.columns,

    [
        "quantity of coal production",
        "coal production quantity",
        "coal production",
        "production quantity",
        "production",
        "quantity"
    ]
)


# =============================================================================
# 12. FALLBACK: MOST NUMERIC COLUMN
# =============================================================================

if production_column is None:

    best_column = None

    best_ratio = -1


    for column in df.columns:

        temp = (
            df[column]
            .astype(str)
            .str.replace(
                ",",
                "",
                regex=False
            )
            .str.strip()
        )


        temp = pd.to_numeric(
            temp,
            errors="coerce"
        )


        ratio = (
            temp.notna().mean()
        )


        if ratio > best_ratio:

            best_ratio = ratio

            best_column = column


    production_column = best_column


# =============================================================================
# 13. DETECT MINE COLUMN
# =============================================================================

mine_column = find_column(

    df.columns,

    [
        "name of mine",
        "mine name",
        "mine"
    ],

    exclude=[
        production_column
    ]
)


# =============================================================================
# 14. DETECT COMPANY COLUMN
# =============================================================================

company_column = find_column(

    df.columns,

    [
        "coal company",
        "company name",
        "name of company",
        "company",
        "subsidiary",
        "region"
    ],

    exclude=[
        production_column,
        mine_column
    ]
)


# =============================================================================
# 15. FALLBACK FOR CATEGORICAL COLUMNS
# =============================================================================

available_columns = [

    column

    for column in df.columns

    if column != production_column
]


if mine_column is None:

    if len(available_columns) >= 1:

        mine_column = (
            available_columns[0]
        )


if company_column is None:

    remaining = [

        column

        for column in available_columns

        if column != mine_column
    ]


    if len(remaining) >= 1:

        company_column = (
            remaining[0]
        )


# =============================================================================
# 16. DISPLAY DETECTED COLUMNS
# =============================================================================

print("\n" + "=" * 80)
print("DETECTED COLUMNS")
print("=" * 80)


print(
    "\nMine column:"
)

print(
    mine_column
)


print(
    "\nCompany column:"
)

print(
    company_column
)


print(
    "\nProduction column:"
)

print(
    production_column
)


if production_column is None:

    raise ValueError(
        "Production column could not be detected."
    )


# =============================================================================
# 17. COPY DATA
# =============================================================================

working_df = df.copy()


# =============================================================================
# 18. CLEAN CATEGORICAL SERIES
# =============================================================================

def clean_categorical_series(series):

    series = series.copy()

    series = series.fillna(
        "Unknown"
    )

    series = series.astype(
        str
    )

    series = series.map(
        lambda value:
        value.strip()
    )

    series = series.replace(

        [
            "",
            "nan",
            "NaN",
            "None",
            "none",
            "null",
            "NULL"
        ],

        "Unknown"
    )

    return series


# =============================================================================
# 19. CLEAN MINE COLUMN
# =============================================================================

if mine_column is not None:

    working_df[
        mine_column
    ] = clean_categorical_series(

        working_df[
            mine_column
        ]
    )


# =============================================================================
# 20. CLEAN COMPANY COLUMN
# =============================================================================

if company_column is not None:

    working_df[
        company_column
    ] = clean_categorical_series(

        working_df[
            company_column
        ]
    )


# =============================================================================
# 21. CLEAN PRODUCTION VALUES
# =============================================================================

production_values = (

    working_df[
        production_column
    ]

    .astype(str)

    .str.replace(
        ",",
        "",
        regex=False
    )

    .str.replace(
        "₹",
        "",
        regex=False
    )

    .str.replace(
        r"[^0-9.\-]",
        "",
        regex=True
    )

    .str.strip()
)


working_df[
    production_column
] = pd.to_numeric(

    production_values,

    errors="coerce"
)


# =============================================================================
# 22. REMOVE INVALID PRODUCTION
# =============================================================================

working_df = (

    working_df

    .dropna(
        subset=[
            production_column
        ]
    )

    .copy()
)


# Remove negative production

working_df = (

    working_df[

        working_df[
            production_column
        ] >= 0

    ]

    .copy()
)


working_df.reset_index(
    drop=True,
    inplace=True
)


print(
    "\nValid records:"
)

print(
    len(working_df)
)


# =============================================================================
# 23. PRODUCTION STATISTICS
# =============================================================================

print("\n" + "=" * 80)
print("PRODUCTION STATISTICS")
print("=" * 80)


print(

    working_df[
        production_column
    ].describe()

)


# =============================================================================
# 24. CREATE TARGET CLASSES
# =============================================================================

TARGET_COLUMN = (
    "Production_Performance"
)


q33 = float(

    working_df[
        production_column
    ].quantile(
        1 / 3
    )
)


q66 = float(

    working_df[
        production_column
    ].quantile(
        2 / 3
    )
)


print(
    "\nLow/Medium threshold:"
)

print(
    q33
)


print(
    "\nMedium/High threshold:"
)

print(
    q66
)


# =============================================================================
# 25. CREATE LOW/MEDIUM/HIGH USING QUANTILES
# =============================================================================

try:

    working_df[
        TARGET_COLUMN
    ] = pd.qcut(

        working_df[
            production_column
        ],

        q=3,

        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


except ValueError:

    print(
        "\nDuplicate quantile boundaries detected."
    )

    print(
        "Using ranked production values."
    )


    ranked_values = (

        working_df[
            production_column
        ]

        .rank(
            method="first"
        )
    )


    working_df[
        TARGET_COLUMN
    ] = pd.qcut(

        ranked_values,

        q=3,

        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


# =============================================================================
# 26. TARGET MAPPING
# =============================================================================

class_mapping = {

    "Low": 0,

    "Medium": 1,

    "High": 2
}


reverse_mapping = {

    0: "Low",

    1: "Medium",

    2: "High"
}


working_df[
    "Target"
] = (

    working_df[
        TARGET_COLUMN
    ]

    .astype(str)

    .map(
        class_mapping
    )
)


print(
    "\nClass distribution:"
)

print(

    working_df[
        TARGET_COLUMN
    ].value_counts()

)


# =============================================================================
# 27. SELECT FEATURES
# =============================================================================

feature_columns = []


if mine_column is not None:

    feature_columns.append(
        mine_column
    )


if (
    company_column is not None
    and company_column not in feature_columns
):

    feature_columns.append(
        company_column
    )


# IMPORTANT:
# Production is deliberately NOT used as a predictor.
# Production was used to create the target.
# Using it would create target leakage.


feature_columns = [

    column

    for column in feature_columns

    if column not in [

        production_column,

        TARGET_COLUMN,

        "Target"
    ]
]


feature_columns = list(
    dict.fromkeys(
        feature_columns
    )
)


if len(feature_columns) == 0:

    raise ValueError(
        "No valid predictor columns found."
    )


print(
    "\nFeatures used:"
)

print(
    feature_columns
)


# =============================================================================
# 28. CREATE X AND y
# =============================================================================

X = working_df[
    feature_columns
].copy()


y = working_df[
    "Target"
].copy()


# =============================================================================
# 29. REMOVE DUPLICATE FEATURE COLUMNS
# =============================================================================

X = X.loc[
    :,
    ~X.columns.duplicated()
].copy()


feature_columns = (
    X.columns.tolist()
)


# =============================================================================
# 30. CLEAN FEATURES SAFELY
# =============================================================================

for column_index in range(
    X.shape[1]
):

    series = X.iloc[
        :,
        column_index
    ]


    cleaned = (
        clean_categorical_series(
            series
        )
    )


    X.iloc[
        :,
        column_index
    ] = cleaned


# =============================================================================
# 31. CHECK TARGET
# =============================================================================

print(
    "\nTarget counts:"
)

print(
    y.value_counts()
)


if y.value_counts().min() < 2:

    raise ValueError(
        "One or more classes contain too few samples."
    )


# =============================================================================
# 32. TRAIN / TEST SPLIT
# =============================================================================

X_train, X_test, y_train, y_test = (
    train_test_split(

        X,
        y,

        test_size=0.20,

        random_state=42,

        stratify=y
    )
)


print(
    "\nTraining samples:"
)

print(
    len(X_train)
)


print(
    "\nTesting samples:"
)

print(
    len(X_test)
)


# =============================================================================
# 33. ONE HOT ENCODER FUNCTION
# =============================================================================

def create_encoder():

    try:

        return OneHotEncoder(

            handle_unknown="ignore",

            sparse_output=False
        )

    except TypeError:

        return OneHotEncoder(

            handle_unknown="ignore",

            sparse=False
        )


# =============================================================================
# 34. RANDOM FOREST PREPROCESSOR
# =============================================================================

rf_encoder = (
    create_encoder()
)


rf_preprocessor = ColumnTransformer(

    transformers=[

        (
            "categorical",

            rf_encoder,

            feature_columns
        )

    ],

    remainder="drop"
)


# =============================================================================
# 35. RANDOM FOREST MODEL
# =============================================================================

rf_classifier = RandomForestClassifier(

    n_estimators=300,

    max_depth=12,

    class_weight="balanced",

    random_state=42,

    n_jobs=-1
)


rf_pipeline = Pipeline(

    steps=[

        (
            "preprocessor",
            rf_preprocessor
        ),

        (
            "classifier",
            rf_classifier
        )
    ]
)


# =============================================================================
# 36. TRAIN RANDOM FOREST
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING RANDOM FOREST")
print("=" * 80)


rf_pipeline.fit(
    X_train,
    y_train
)


# =============================================================================
# 37. RANDOM FOREST PREDICTION
# =============================================================================

rf_predictions = (
    rf_pipeline.predict(
        X_test
    )
)


# =============================================================================
# 38. RANDOM FOREST METRICS
# =============================================================================

rf_accuracy = accuracy_score(
    y_test,
    rf_predictions
)


rf_precision = precision_score(
    y_test,
    rf_predictions,
    average="weighted",
    zero_division=0
)


rf_recall = recall_score(
    y_test,
    rf_predictions,
    average="weighted",
    zero_division=0
)


rf_f1 = f1_score(
    y_test,
    rf_predictions,
    average="weighted",
    zero_division=0
)


print(
    "\nRandom Forest Accuracy:"
)

print(
    f"{rf_accuracy * 100:.2f}%"
)


print(
    "\nClassification Report:"
)


print(

    classification_report(

        y_test,

        rf_predictions,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Low",
            "Medium",
            "High"
        ],

        zero_division=0
    )

)


# =============================================================================
# 39. NEURAL NETWORK PREPROCESSOR
# =============================================================================

nn_encoder = (
    create_encoder()
)


nn_preprocessor = ColumnTransformer(

    transformers=[

        (
            "categorical",

            nn_encoder,

            feature_columns
        )

    ],

    remainder="drop"
)


# =============================================================================
# 40. TRANSFORM NN FEATURES
# =============================================================================

X_train_nn = (
    nn_preprocessor.fit_transform(
        X_train
    )
)


X_test_nn = (
    nn_preprocessor.transform(
        X_test
    )
)


X_train_nn = np.asarray(
    X_train_nn,
    dtype=np.float32
)


X_test_nn = np.asarray(
    X_test_nn,
    dtype=np.float32
)


# =============================================================================
# 41. SCALE NN FEATURES
# =============================================================================

scaler = StandardScaler()


X_train_nn = scaler.fit_transform(
    X_train_nn
)


X_test_nn = scaler.transform(
    X_test_nn
)


X_train_nn = np.asarray(
    X_train_nn,
    dtype=np.float32
)


X_test_nn = np.asarray(
    X_test_nn,
    dtype=np.float32
)


# =============================================================================
# 42. CREATE NEURAL NETWORK
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING NEURAL NETWORK")
print("=" * 80)


input_dimension = (
    X_train_nn.shape[1]
)


nn_model = Sequential(
    [

        Input(
            shape=(
                input_dimension,
            )
        ),

        Dense(
            128,
            activation="relu"
        ),

        Dropout(
            0.30
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dropout(
            0.20
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            3,
            activation="softmax"
        )

    ]
)


nn_model.compile(

    optimizer="adam",

    loss="sparse_categorical_crossentropy",

    metrics=[
        "accuracy"
    ]
)


# =============================================================================
# 43. EARLY STOPPING
# =============================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=15,

    restore_best_weights=True,

    verbose=1
)


# =============================================================================
# 44. TRAIN NN
# =============================================================================

history = nn_model.fit(

    X_train_nn,

    y_train.to_numpy(
        dtype=np.int32
    ),

    validation_split=0.20,

    epochs=150,

    batch_size=16,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# =============================================================================
# 45. NN PREDICTIONS
# =============================================================================

nn_probabilities = (
    nn_model.predict(
        X_test_nn,
        verbose=0
    )
)


nn_predictions = np.argmax(
    nn_probabilities,
    axis=1
)


# =============================================================================
# 46. NN METRICS
# =============================================================================

nn_accuracy = accuracy_score(
    y_test,
    nn_predictions
)


nn_precision = precision_score(
    y_test,
    nn_predictions,
    average="weighted",
    zero_division=0
)


nn_recall = recall_score(
    y_test,
    nn_predictions,
    average="weighted",
    zero_division=0
)


nn_f1 = f1_score(
    y_test,
    nn_predictions,
    average="weighted",
    zero_division=0
)


print(
    "\nNeural Network Accuracy:"
)

print(
    f"{nn_accuracy * 100:.2f}%"
)


print(
    "\nNeural Network Classification Report:"
)


print(

    classification_report(

        y_test,

        nn_predictions,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Low",
            "Medium",
            "High"
        ],

        zero_division=0
    )

)


# =============================================================================
# 47. SAVE PROCESSED DATA
# =============================================================================

working_df.to_csv(

    PROCESSED_DATA_PATH,

    index=False
)


print(
    "\nProcessed dataset saved:"
)

print(
    PROCESSED_DATA_PATH
)


# =============================================================================
# 48. ACCURACY GRAPH
# =============================================================================

print("\n" + "=" * 80)
print("CREATING ACCURACY GRAPH")
print("=" * 80)


train_accuracy = (
    history.history[
        "accuracy"
    ]
)


validation_accuracy = (
    history.history[
        "val_accuracy"
    ]
)


epochs = range(
    1,
    len(train_accuracy) + 1
)


plt.figure(
    figsize=(10, 6)
)


plt.plot(
    epochs,
    train_accuracy,
    marker="o",
    markersize=3,
    label="Training Accuracy"
)


plt.plot(
    epochs,
    validation_accuracy,
    marker="s",
    markersize=3,
    label="Validation Accuracy"
)


plt.xlabel(
    "Epoch"
)


plt.ylabel(
    "Accuracy"
)


plt.title(
    "Neural Network Training and Validation Accuracy"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    ACCURACY_GRAPH_PATH
)


# =============================================================================
# 49. CREATE PREDICTION CSV
# =============================================================================

print("\n" + "=" * 80)
print("CREATING PREDICTION CSV")
print("=" * 80)


prediction_df = (
    X_test
    .reset_index(
        drop=True
    )
    .copy()
)


actual_array = np.asarray(
    y_test
)


rf_array = np.asarray(
    rf_predictions
)


nn_array = np.asarray(
    nn_predictions
)


prediction_df[
    "Actual_Class_Code"
] = actual_array


prediction_df[
    "Actual_Class"
] = [

    reverse_mapping[
        int(value)
    ]

    for value in actual_array
]


prediction_df[
    "Random_Forest_Prediction_Code"
] = rf_array


prediction_df[
    "Random_Forest_Prediction"
] = [

    reverse_mapping[
        int(value)
    ]

    for value in rf_array
]


prediction_df[
    "Neural_Network_Prediction_Code"
] = nn_array


prediction_df[
    "Neural_Network_Prediction"
] = [

    reverse_mapping[
        int(value)
    ]

    for value in nn_array
]


prediction_df[
    "RF_Correct"
] = (
    actual_array
    == rf_array
)


prediction_df[
    "NN_Correct"
] = (
    actual_array
    == nn_array
)


# =============================================================================
# 50. ADD NN PROBABILITIES
# =============================================================================

prediction_df[
    "NN_Probability_Low"
] = nn_probabilities[
    :,
    0
]


prediction_df[
    "NN_Probability_Medium"
] = nn_probabilities[
    :,
    1
]


prediction_df[
    "NN_Probability_High"
] = nn_probabilities[
    :,
    2
]


# =============================================================================
# 51. SAVE PREDICTION CSV
# =============================================================================

prediction_df.to_csv(

    PREDICTION_CSV_PATH,

    index=False
)


print(
    "\nPrediction CSV saved:"
)

print(
    PREDICTION_CSV_PATH
)


print(
    "\nPrediction Preview:"
)

print(
    prediction_df.head(10)
)


# =============================================================================
# 52. PREDICTION GRAPH
# =============================================================================

print("\n" + "=" * 80)
print("CREATING PREDICTION GRAPH")
print("=" * 80)


number_of_samples = min(
    50,
    len(actual_array)
)


samples = np.arange(
    1,
    number_of_samples + 1
)


plt.figure(
    figsize=(14, 7)
)


plt.plot(

    samples,

    actual_array[
        :number_of_samples
    ],

    marker="o",

    linewidth=2,

    label="Actual"
)


plt.plot(

    samples,

    rf_array[
        :number_of_samples
    ],

    marker="s",

    linestyle="--",

    linewidth=1.8,

    label="Random Forest"
)


plt.plot(

    samples,

    nn_array[
        :number_of_samples
    ],

    marker="^",

    linestyle=":",

    linewidth=1.8,

    label="Neural Network"
)


plt.xlabel(
    "Test Sample"
)


plt.ylabel(
    "Production Performance"
)


plt.title(
    "Actual vs Predicted Coal Mine Performance"
)


plt.yticks(

    [
        0,
        1,
        2
    ],

    [
        "Low",
        "Medium",
        "High"
    ]
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    PREDICTION_GRAPH_PATH
)


# =============================================================================
# 53. CREATE RESULT CSV
# =============================================================================

print("\n" + "=" * 80)
print("CREATING RESULT CSV")
print("=" * 80)


result_df = pd.DataFrame(
    {

        "Model": [

            "Random Forest",

            "Neural Network"
        ],


        "Accuracy": [

            rf_accuracy,

            nn_accuracy
        ],


        "Accuracy_Percentage": [

            rf_accuracy * 100,

            nn_accuracy * 100
        ],


        "Precision": [

            rf_precision,

            nn_precision
        ],


        "Precision_Percentage": [

            rf_precision * 100,

            nn_precision * 100
        ],


        "Recall": [

            rf_recall,

            nn_recall
        ],


        "Recall_Percentage": [

            rf_recall * 100,

            nn_recall * 100
        ],


        "F1_Score": [

            rf_f1,

            nn_f1
        ],


        "F1_Percentage": [

            rf_f1 * 100,

            nn_f1 * 100
        ]
    }
)


result_df.to_csv(

    RESULT_CSV_PATH,

    index=False
)


print(
    "\nResult CSV saved:"
)

print(
    RESULT_CSV_PATH
)


print(
    "\nResults:"
)

print(
    result_df
)


# =============================================================================
# 54. RESULT GRAPH
# =============================================================================

print("\n" + "=" * 80)
print("CREATING RESULT GRAPH")
print("=" * 80)


models = [
    "Random Forest",
    "Neural Network"
]


model_accuracy = [

    rf_accuracy * 100,

    nn_accuracy * 100
]


plt.figure(
    figsize=(9, 6)
)


bars = plt.bar(
    models,
    model_accuracy
)


plt.xlabel(
    "Model"
)


plt.ylabel(
    "Accuracy (%)"
)


plt.title(
    "Coal Mine Performance Classification Results"
)


plt.ylim(
    0,
    100
)


for bar, value in zip(
    bars,
    model_accuracy
):

    plt.text(

        bar.get_x()
        + bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center",

        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    RESULT_GRAPH_PATH
)


# =============================================================================
# 55. COMPARISON GRAPH
# =============================================================================

print("\n" + "=" * 80)
print("CREATING COMPARISON GRAPH")
print("=" * 80)


metric_names = [

    "Accuracy",

    "Precision",

    "Recall",

    "F1 Score"
]


rf_scores = [

    rf_accuracy * 100,

    rf_precision * 100,

    rf_recall * 100,

    rf_f1 * 100
]


nn_scores = [

    nn_accuracy * 100,

    nn_precision * 100,

    nn_recall * 100,

    nn_f1 * 100
]


x = np.arange(
    len(metric_names)
)


width = 0.35


plt.figure(
    figsize=(11, 7)
)


rf_bars = plt.bar(

    x - width / 2,

    rf_scores,

    width,

    label="Random Forest"
)


nn_bars = plt.bar(

    x + width / 2,

    nn_scores,

    width,

    label="Neural Network"
)


plt.xlabel(
    "Evaluation Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "Random Forest vs Neural Network"
)


plt.xticks(
    x,
    metric_names
)


plt.ylim(
    0,
    100
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


# =============================================================================
# 56. VALUES ON COMPARISON BARS
# =============================================================================

for bars in [
    rf_bars,
    nn_bars
]:

    for bar in bars:

        height = (
            bar.get_height()
        )


        plt.text(

            bar.get_x()
            + bar.get_width() / 2,

            height + 0.7,

            f"{height:.1f}",

            ha="center",

            fontsize=9
        )


plt.tight_layout()


plt.savefig(

    COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    COMPARISON_GRAPH_PATH
)


# =============================================================================
# 57. HEATMAP
# =============================================================================

print("\n" + "=" * 80)
print("CREATING CONFUSION MATRIX HEATMAP")
print("=" * 80)


# Use model with highest accuracy

if rf_accuracy >= nn_accuracy:

    heatmap_predictions = (
        rf_predictions
    )

    heatmap_model = (
        "Random Forest"
    )

else:

    heatmap_predictions = (
        nn_predictions
    )

    heatmap_model = (
        "Neural Network"
    )


cm = confusion_matrix(

    y_test,

    heatmap_predictions,

    labels=[
        0,
        1,
        2
    ]
)


print(
    "\nHeatmap model:"
)

print(
    heatmap_model
)


print(
    "\nConfusion Matrix:"
)

print(
    cm
)


# =============================================================================
# 58. DRAW HEATMAP WITHOUT SEABORN
# =============================================================================

class_names = [

    "Low",

    "Medium",

    "High"
]


plt.figure(
    figsize=(8, 7)
)


image = plt.imshow(

    cm,

    interpolation="nearest",

    cmap="Blues"
)


plt.colorbar(
    image
)


plt.title(
    f"Confusion Matrix - {heatmap_model}"
)


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "Actual Class"
)


ticks = np.arange(
    len(class_names)
)


plt.xticks(
    ticks,
    class_names
)


plt.yticks(
    ticks,
    class_names
)


# =============================================================================
# 59. ADD VALUES INSIDE HEATMAP
# =============================================================================

threshold = (

    cm.max() / 2

    if cm.size > 0

    else 0
)


for row in range(
    cm.shape[0]
):

    for column in range(
        cm.shape[1]
    ):

        value = cm[
            row,
            column
        ]


        if value > threshold:

            text_color = (
                "white"
            )

        else:

            text_color = (
                "black"
            )


        plt.text(

            column,

            row,

            str(value),

            ha="center",

            va="center",

            color=text_color,

            fontsize=14,

            fontweight="bold"
        )


plt.tight_layout()


plt.savefig(

    HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    HEATMAP_PATH
)


# =============================================================================
# 60. VERIFY ALL REQUIRED OUTPUTS
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING GENERATED FILES")
print("=" * 80)


required_files = [

    ACCURACY_GRAPH_PATH,

    PREDICTION_CSV_PATH,

    PREDICTION_GRAPH_PATH,

    RESULT_CSV_PATH,

    RESULT_GRAPH_PATH,

    COMPARISON_GRAPH_PATH,

    HEATMAP_PATH,

    PROCESSED_DATA_PATH
]


for file_path in required_files:

    if os.path.exists(
        file_path
    ):

        size = os.path.getsize(
            file_path
        )


        print(

            f"[OK] "
            f"{os.path.basename(file_path)} "
            f"({size:,} bytes)"
        )

    else:

        print(

            f"[MISSING] "
            f"{os.path.basename(file_path)}"
        )


# =============================================================================
# 61. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 80)
print("FINAL PROJECT RESULTS")
print("=" * 80)


print(
    f"""

AI-Based Coal Mine Production Analysis
and Performance Classification


RANDOM FOREST
=============

Accuracy:
{rf_accuracy * 100:.2f}%

Precision:
{rf_precision * 100:.2f}%

Recall:
{rf_recall * 100:.2f}%

F1 Score:
{rf_f1 * 100:.2f}%


NEURAL NETWORK
==============

Accuracy:
{nn_accuracy * 100:.2f}%

Precision:
{nn_precision * 100:.2f}%

Recall:
{nn_recall * 100:.2f}%

F1 Score:
{nn_f1 * 100:.2f}%


GENERATED FILES
===============

1. accuracy_graph.png

2. prediction.csv

3. prediction_graph.png

4. result.csv

5. result_graph.png

6. comparison_graph.png

7. heatmap.png

8. processed_coal_production_data.csv


ALL FILES SAVED IN
==================

{PROJECT_DIR}

"""
)


print("=" * 80)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 80)
```
