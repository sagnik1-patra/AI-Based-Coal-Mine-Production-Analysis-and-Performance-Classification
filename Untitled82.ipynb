```python
import os
import re
import json
import yaml
import pickle
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.preprocessing import OneHotEncoder

from sklearn.compose import ColumnTransformer

from sklearn.pipeline import Pipeline

from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)


# =============================================================================
# 2. GENERAL SETTINGS
# =============================================================================

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

np.random.seed(
    RANDOM_STATE
)

random.seed(
    RANDOM_STATE
)


# =============================================================================
# 3. PROJECT DIRECTORY
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Coal Mine Production Analysis and Performance Classification"
)


CSV_PATH = os.path.join(
    PROJECT_DIR,
    "rs_session-241_au482_1.1.csv"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 4. OUTPUT PATHS
# =============================================================================

PSO_MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "pso_model.pkl"
)


PSO_CONFIG_PATH = os.path.join(
    PROJECT_DIR,
    "pso_config.yaml"
)


PSO_METADATA_PATH = os.path.join(
    PROJECT_DIR,
    "pso_metadata.json"
)


PSO_PROCESSED_PATH = os.path.join(
    PROJECT_DIR,
    "pso_processed_data.csv"
)


PSO_PREDICTION_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction.csv"
)


PSO_RESULT_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result.csv"
)


PSO_OPTIMIZATION_HISTORY_PATH = os.path.join(
    PROJECT_DIR,
    "pso_optimization_history.csv"
)


# =============================================================================
# 5. GRAPH PATHS
# =============================================================================

PSO_ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_accuracy_graph.png"
)


PSO_PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_prediction_graph.png"
)


PSO_RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_result_graph.png"
)


PSO_COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_comparison_graph.png"
)


PSO_HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "pso_heatmap.png"
)


PSO_OPTIMIZATION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "pso_optimization_graph.png"
)


PSO_CLASS_DISTRIBUTION_PATH = os.path.join(
    PROJECT_DIR,
    "pso_class_distribution_graph.png"
)


# =============================================================================
# 6. START
# =============================================================================

print("=" * 90)

print(
    "AI-BASED COAL MINE PRODUCTION ANALYSIS"
)

print(
    "PARTICLE SWARM OPTIMIZATION (PSO)"
)

print("=" * 90)


# =============================================================================
# 7. CHECK DATASET
# =============================================================================

if not os.path.exists(
    CSV_PATH
):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}"
    )


print(
    "\nDataset found:"
)

print(
    CSV_PATH
)


# =============================================================================
# 8. LOAD DATASET
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "LOADING DATASET"
)

print(
    "=" * 90
)


encodings = [
    "utf-8",
    "utf-8-sig",
    "cp1252",
    "latin1"
]


df = None

used_encoding = None


for encoding in encodings:

    try:

        df = pd.read_csv(
            CSV_PATH,
            encoding=encoding
        )

        used_encoding = (
            encoding
        )

        break

    except UnicodeDecodeError:

        continue


if df is None:

    raise ValueError(
        "Could not read the CSV file."
    )


print(
    "\nDataset loaded successfully."
)

print(
    "Encoding:",
    used_encoding
)

print(
    "Shape:",
    df.shape
)


# =============================================================================
# 9. CLEAN COLUMN NAMES
# =============================================================================

def clean_column_name(
    column
):

    column = str(
        column
    )

    column = column.replace(
        "\n",
        " "
    )

    column = column.replace(
        "\r",
        " "
    )

    column = re.sub(
        r"\s+",
        " ",
        column
    )

    return column.strip()


df.columns = [

    clean_column_name(
        column
    )

    for column
    in df.columns
]


# =============================================================================
# 10. MAKE COLUMN NAMES UNIQUE
# =============================================================================

def make_unique_columns(
    columns
):

    counts = {}

    output = []


    for column in columns:

        if column not in counts:

            counts[
                column
            ] = 0

            output.append(
                column
            )

        else:

            counts[
                column
            ] += 1


            new_name = (
                f"{column}_"
                f"{counts[column]}"
            )


            while new_name in output:

                counts[
                    column
                ] += 1

                new_name = (
                    f"{column}_"
                    f"{counts[column]}"
                )


            output.append(
                new_name
            )


    return output


df.columns = (
    make_unique_columns(
        df.columns
    )
)


print(
    "\nColumns:"
)


for index, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{index}. {column}"
    )


# =============================================================================
# 11. REMOVE EMPTY ROWS AND COLUMNS
# =============================================================================

df = df.dropna(
    axis=0,
    how="all"
)


df = df.dropna(
    axis=1,
    how="all"
)


# =============================================================================
# 12. COLUMN SEARCH FUNCTION
# =============================================================================

def find_column(
    columns,
    keywords,
    exclude=None
):

    if exclude is None:

        exclude = []


    # -------------------------------------------------------------------------
    # Exact match
    # -------------------------------------------------------------------------

    for keyword in keywords:

        for column in columns:

            if column in exclude:

                continue


            if (
                str(column)
                .lower()
                .strip()
                ==
                keyword
                .lower()
                .strip()
            ):

                return column


    # -------------------------------------------------------------------------
    # Partial match
    # -------------------------------------------------------------------------

    for keyword in keywords:

        for column in columns:

            if column in exclude:

                continue


            if (
                keyword.lower()
                in
                str(column).lower()
            ):

                return column


    return None


# =============================================================================
# 13. DETECT PRODUCTION COLUMN
# =============================================================================

production_column = find_column(

    df.columns,

    [
        "quantity of coal production",
        "coal production quantity",
        "coal production",
        "production quantity",
        "production",
        "quantity"
    ]
)


# =============================================================================
# 14. PRODUCTION COLUMN FALLBACK
# =============================================================================

if production_column is None:

    best_column = None

    best_ratio = -1


    for column in df.columns:

        values = (

            df[column]

            .astype(str)

            .str.replace(
                ",",
                "",
                regex=False
            )

            .str.strip()
        )


        numeric_values = (
            pd.to_numeric(
                values,
                errors="coerce"
            )
        )


        ratio = (
            numeric_values
            .notna()
            .mean()
        )


        if ratio > best_ratio:

            best_ratio = (
                ratio
            )

            best_column = (
                column
            )


    production_column = (
        best_column
    )


# =============================================================================
# 15. DETECT MINE COLUMN
# =============================================================================

mine_column = find_column(

    df.columns,

    [
        "name of mine",
        "mine name",
        "mine"
    ],

    exclude=[
        production_column
    ]
)


# =============================================================================
# 16. DETECT COMPANY COLUMN
# =============================================================================

company_column = find_column(

    df.columns,

    [
        "coal company",
        "company name",
        "name of company",
        "company",
        "subsidiary",
        "region"
    ],

    exclude=[
        production_column,
        mine_column
    ]
)


# =============================================================================
# 17. FALLBACK CATEGORICAL COLUMNS
# =============================================================================

available_columns = [

    column

    for column in df.columns

    if column != production_column
]


if mine_column is None:

    if len(
        available_columns
    ) >= 1:

        mine_column = (
            available_columns[0]
        )


if company_column is None:

    remaining = [

        column

        for column in available_columns

        if column != mine_column
    ]


    if len(
        remaining
    ) >= 1:

        company_column = (
            remaining[0]
        )


# =============================================================================
# 18. DISPLAY DETECTED COLUMNS
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "DETECTED COLUMNS"
)

print(
    "=" * 90
)


print(
    "\nMine column:"
)

print(
    mine_column
)


print(
    "\nCompany column:"
)

print(
    company_column
)


print(
    "\nProduction column:"
)

print(
    production_column
)


if production_column is None:

    raise ValueError(
        "Production column could not be detected."
    )


# =============================================================================
# 19. COPY DATA
# =============================================================================

working_df = (
    df.copy()
)


# =============================================================================
# 20. CLEAN CATEGORICAL SERIES
# =============================================================================

def clean_categorical_series(
    series
):

    series = (
        series.copy()
    )


    series = (
        series.fillna(
            "Unknown"
        )
    )


    series = (
        series.astype(
            str
        )
    )


    series = (
        series.map(
            lambda value:
            value.strip()
        )
    )


    series = (
        series.replace(

            [
                "",
                "nan",
                "NaN",
                "None",
                "none",
                "NULL",
                "null"
            ],

            "Unknown"
        )
    )


    return series


# =============================================================================
# 21. CLEAN MINE
# =============================================================================

if mine_column is not None:

    working_df[
        mine_column
    ] = clean_categorical_series(

        working_df[
            mine_column
        ]
    )


# =============================================================================
# 22. CLEAN COMPANY
# =============================================================================

if company_column is not None:

    working_df[
        company_column
    ] = clean_categorical_series(

        working_df[
            company_column
        ]
    )


# =============================================================================
# 23. CLEAN PRODUCTION
# =============================================================================

production_values = (

    working_df[
        production_column
    ]

    .astype(str)

    .str.replace(
        ",",
        "",
        regex=False
    )

    .str.replace(
        "₹",
        "",
        regex=False
    )

    .str.replace(
        r"[^0-9.\-]",
        "",
        regex=True
    )

    .str.strip()
)


working_df[
    production_column
] = pd.to_numeric(

    production_values,

    errors="coerce"
)


# =============================================================================
# 24. REMOVE INVALID PRODUCTION
# =============================================================================

working_df = (

    working_df

    .dropna(
        subset=[
            production_column
        ]
    )

    .copy()
)


working_df = (

    working_df[

        working_df[
            production_column
        ] >= 0

    ]

    .copy()
)


working_df.reset_index(
    drop=True,
    inplace=True
)


if len(
    working_df
) < 10:

    raise ValueError(
        "Too few valid records after preprocessing."
    )


print(
    "\nValid records:",
    len(working_df)
)


# =============================================================================
# 25. PRODUCTION STATISTICS
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "PRODUCTION STATISTICS"
)

print(
    "=" * 90
)


print(

    working_df[
        production_column
    ].describe()

)


# =============================================================================
# 26. CREATE TARGET CLASS
# =============================================================================

TARGET_COLUMN = (
    "Production_Performance"
)


q33 = float(

    working_df[
        production_column
    ]

    .quantile(
        1 / 3
    )
)


q66 = float(

    working_df[
        production_column
    ]

    .quantile(
        2 / 3
    )
)


print(
    "\nLow / Medium threshold:",
    q33
)


print(
    "Medium / High threshold:",
    q66
)


# =============================================================================
# 27. CREATE LOW/MEDIUM/HIGH CLASSES
# =============================================================================

try:

    working_df[
        TARGET_COLUMN
    ] = pd.qcut(

        working_df[
            production_column
        ],

        q=3,

        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


except ValueError:

    print(
        "\nDuplicate quantile boundaries found."
    )

    print(
        "Using ranked production values."
    )


    ranked_values = (

        working_df[
            production_column
        ]

        .rank(
            method="first"
        )
    )


    working_df[
        TARGET_COLUMN
    ] = pd.qcut(

        ranked_values,

        q=3,

        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


# =============================================================================
# 28. TARGET MAPPING
# =============================================================================

class_mapping = {

    "Low": 0,

    "Medium": 1,

    "High": 2
}


reverse_mapping = {

    0: "Low",

    1: "Medium",

    2: "High"
}


working_df[
    "Target"
] = (

    working_df[
        TARGET_COLUMN
    ]

    .astype(str)

    .map(
        class_mapping
    )
)


print(
    "\nClass distribution:"
)


print(

    working_df[
        TARGET_COLUMN
    ].value_counts()

)


# =============================================================================
# 29. SELECT FEATURES
# =============================================================================

feature_columns = []


if mine_column is not None:

    feature_columns.append(
        mine_column
    )


if (
    company_column is not None
    and
    company_column not in feature_columns
):

    feature_columns.append(
        company_column
    )


# =============================================================================
# IMPORTANT:
#
# The production column created the target class.
# Therefore production itself must NOT be used as an input feature.
#
# Otherwise the model would suffer from target leakage.
# =============================================================================

feature_columns = [

    column

    for column
    in feature_columns

    if column not in [

        production_column,

        TARGET_COLUMN,

        "Target"
    ]
]


feature_columns = list(
    dict.fromkeys(
        feature_columns
    )
)


if len(
    feature_columns
) == 0:

    raise ValueError(
        "No valid predictor columns were found."
    )


print(
    "\nFeatures used:"
)

print(
    feature_columns
)


# =============================================================================
# 30. CREATE X AND y
# =============================================================================

X = (

    working_df[
        feature_columns
    ]

    .copy()
)


y = (

    working_df[
        "Target"
    ]

    .copy()
)


# =============================================================================
# 31. REMOVE DUPLICATE FEATURE COLUMNS
# =============================================================================

X = X.loc[
    :,
    ~X.columns.duplicated()
].copy()


feature_columns = (
    X.columns.tolist()
)


# =============================================================================
# 32. CLEAN FEATURES
# =============================================================================

for column_index in range(
    X.shape[1]
):

    X.iloc[
        :,
        column_index
    ] = clean_categorical_series(

        X.iloc[
            :,
            column_index
        ]
    )


# =============================================================================
# 33. CHECK CLASS COUNTS
# =============================================================================

print(
    "\nTarget counts:"
)


print(
    y.value_counts()
)


if y.value_counts().min() < 2:

    raise ValueError(
        "One or more classes contain too few samples."
    )


# =============================================================================
# 34. TRAIN TEST SPLIT
# =============================================================================

X_train, X_test, y_train, y_test = (
    train_test_split(

        X,

        y,

        test_size=0.20,

        random_state=RANDOM_STATE,

        stratify=y
    )
)


print(
    "\nTraining samples:",
    len(X_train)
)


print(
    "Testing samples:",
    len(X_test)
)


# =============================================================================
# 35. CREATE ONE HOT ENCODER
# =============================================================================

def create_encoder():

    try:

        return OneHotEncoder(

            handle_unknown="ignore",

            sparse_output=True
        )

    except TypeError:

        return OneHotEncoder(

            handle_unknown="ignore",

            sparse=True
        )


# =============================================================================
# 36. PSO SEARCH SPACE
# =============================================================================
#
# Particle position:
#
# dimension 0 -> n_estimators
# dimension 1 -> max_depth
# dimension 2 -> min_samples_split
# dimension 3 -> min_samples_leaf
# dimension 4 -> max_features option
#
# max_features:
#
# 0 = sqrt
# 1 = log2
# 2 = None
#
# =============================================================================

LOWER_BOUNDS = np.array(
    [
        100,
        3,
        2,
        1,
        0
    ],
    dtype=float
)


UPPER_BOUNDS = np.array(
    [
        500,
        25,
        15,
        10,
        2
    ],
    dtype=float
)


MAX_FEATURE_OPTIONS = [

    "sqrt",

    "log2",

    None
]


# =============================================================================
# 37. PSO SETTINGS
# =============================================================================

PSO_SWARM_SIZE = 10

PSO_ITERATIONS = 10


# Inertia

PSO_W_START = 0.90

PSO_W_END = 0.40


# Cognitive coefficient

PSO_C1 = 1.7


# Social coefficient

PSO_C2 = 1.7


print(
    "\n" + "=" * 90
)

print(
    "PSO CONFIGURATION"
)

print(
    "=" * 90
)


print(
    "Swarm size:",
    PSO_SWARM_SIZE
)


print(
    "Iterations:",
    PSO_ITERATIONS
)


print(
    "Initial inertia:",
    PSO_W_START
)


print(
    "Final inertia:",
    PSO_W_END
)


print(
    "Cognitive coefficient:",
    PSO_C1
)


print(
    "Social coefficient:",
    PSO_C2
)


# =============================================================================
# 38. PARTICLE POSITION -> RF PARAMETERS
# =============================================================================

def position_to_parameters(
    position
):

    n_estimators = int(
        round(
            position[0]
        )
    )


    max_depth = int(
        round(
            position[1]
        )
    )


    min_samples_split = int(
        round(
            position[2]
        )
    )


    min_samples_leaf = int(
        round(
            position[3]
        )
    )


    max_features_index = int(
        round(
            position[4]
        )
    )


    # Enforce bounds

    n_estimators = int(
        np.clip(
            n_estimators,
            100,
            500
        )
    )


    max_depth = int(
        np.clip(
            max_depth,
            3,
            25
        )
    )


    min_samples_split = int(
        np.clip(
            min_samples_split,
            2,
            15
        )
    )


    min_samples_leaf = int(
        np.clip(
            min_samples_leaf,
            1,
            10
        )
    )


    max_features_index = int(
        np.clip(
            max_features_index,
            0,
            2
        )
    )


    parameters = {

        "n_estimators":
            n_estimators,

        "max_depth":
            max_depth,

        "min_samples_split":
            min_samples_split,

        "min_samples_leaf":
            min_samples_leaf,

        "max_features":
            MAX_FEATURE_OPTIONS[
                max_features_index
            ]
    }


    return parameters


# =============================================================================
# 39. CREATE RANDOM FOREST PIPELINE
# =============================================================================

def create_pipeline(
    parameters
):

    encoder = (
        create_encoder()
    )


    preprocessor = ColumnTransformer(

        transformers=[

            (
                "categorical",

                encoder,

                feature_columns
            )

        ],

        remainder="drop"
    )


    classifier = RandomForestClassifier(

        n_estimators=int(
            parameters[
                "n_estimators"
            ]
        ),

        max_depth=int(
            parameters[
                "max_depth"
            ]
        ),

        min_samples_split=int(
            parameters[
                "min_samples_split"
            ]
        ),

        min_samples_leaf=int(
            parameters[
                "min_samples_leaf"
            ]
        ),

        max_features=parameters[
            "max_features"
        ],

        class_weight="balanced",

        random_state=RANDOM_STATE,

        n_jobs=-1
    )


    pipeline = Pipeline(

        steps=[

            (
                "preprocessor",
                preprocessor
            ),

            (
                "classifier",
                classifier
            )
        ]
    )


    return pipeline


# =============================================================================
# 40. CROSS-VALIDATION
# =============================================================================

minimum_class_count = int(
    y_train.value_counts().min()
)


if minimum_class_count >= 3:

    CV_SPLITS = 3

else:

    CV_SPLITS = 2


cv = StratifiedKFold(

    n_splits=CV_SPLITS,

    shuffle=True,

    random_state=RANDOM_STATE
)


print(
    "\nCross-validation folds:",
    CV_SPLITS
)


# =============================================================================
# 41. FITNESS CACHE
# =============================================================================

fitness_cache = {}


# =============================================================================
# 42. FITNESS FUNCTION
# =============================================================================

def parameter_key(
    parameters
):

    return (

        parameters[
            "n_estimators"
        ],

        parameters[
            "max_depth"
        ],

        parameters[
            "min_samples_split"
        ],

        parameters[
            "min_samples_leaf"
        ],

        str(
            parameters[
                "max_features"
            ]
        )
    )


def evaluate_particle(
    position
):

    parameters = (
        position_to_parameters(
            position
        )
    )


    key = (
        parameter_key(
            parameters
        )
    )


    if key in fitness_cache:

        return (
            fitness_cache[
                key
            ]
        )


    model = (
        create_pipeline(
            parameters
        )
    )


    scores = cross_val_score(

        model,

        X_train,

        y_train,

        cv=cv,

        scoring="accuracy",

        n_jobs=1
    )


    fitness = float(
        np.mean(
            scores
        )
    )


    fitness_cache[
        key
    ] = fitness


    return fitness


# =============================================================================
# 43. INITIALIZE PARTICLES
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "INITIALIZING PSO SWARM"
)

print(
    "=" * 90
)


dimensions = len(
    LOWER_BOUNDS
)


positions = np.random.uniform(

    low=LOWER_BOUNDS,

    high=UPPER_BOUNDS,

    size=(
        PSO_SWARM_SIZE,
        dimensions
    )
)


# =============================================================================
# 44. INITIALIZE VELOCITIES
# =============================================================================

velocity_range = (
    UPPER_BOUNDS
    -
    LOWER_BOUNDS
)


velocities = np.random.uniform(

    low=(
        -0.10
        *
        velocity_range
    ),

    high=(
        0.10
        *
        velocity_range
    ),

    size=(
        PSO_SWARM_SIZE,
        dimensions
    )
)


# =============================================================================
# 45. PERSONAL BESTS
# =============================================================================

personal_best_positions = (
    positions.copy()
)


personal_best_scores = np.full(
    PSO_SWARM_SIZE,
    -np.inf
)


# =============================================================================
# 46. GLOBAL BEST
# =============================================================================

global_best_position = None

global_best_score = (
    -np.inf
)


# =============================================================================
# 47. PSO HISTORY
# =============================================================================

optimization_history = []


# =============================================================================
# 48. START PSO
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "STARTING PARTICLE SWARM OPTIMIZATION"
)

print(
    "=" * 90
)


for iteration in range(
    1,
    PSO_ITERATIONS + 1
):

    print(
        f"\nIteration "
        f"{iteration}/"
        f"{PSO_ITERATIONS}"
    )


    print(
        "-" * 90
    )


    iteration_scores = []


    # =========================================================================
    # 49. EVALUATE PARTICLES
    # =========================================================================

    for particle_index in range(
        PSO_SWARM_SIZE
    ):

        position = (
            positions[
                particle_index
            ]
        )


        fitness = (
            evaluate_particle(
                position
            )
        )


        iteration_scores.append(
            fitness
        )


        parameters = (
            position_to_parameters(
                position
            )
        )


        print(
            f"Particle "
            f"{particle_index + 1:02d} | "
            f"Fitness = "
            f"{fitness:.4f} | "
            f"{parameters}"
        )


        # =====================================================================
        # PERSONAL BEST
        # =====================================================================

        if (
            fitness
            >
            personal_best_scores[
                particle_index
            ]
        ):

            personal_best_scores[
                particle_index
            ] = fitness


            personal_best_positions[
                particle_index
            ] = (
                position.copy()
            )


        # =====================================================================
        # GLOBAL BEST
        # =====================================================================

        if (
            fitness
            >
            global_best_score
        ):

            global_best_score = (
                fitness
            )


            global_best_position = (
                position.copy()
            )


    # =========================================================================
    # 50. ITERATION STATISTICS
    # =========================================================================

    iteration_best = float(
        np.max(
            iteration_scores
        )
    )


    iteration_average = float(
        np.mean(
            iteration_scores
        )
    )


    iteration_worst = float(
        np.min(
            iteration_scores
        )
    )


    global_parameters = (
        position_to_parameters(
            global_best_position
        )
    )


    # =========================================================================
    # INERTIA WEIGHT
    # =========================================================================

    if PSO_ITERATIONS > 1:

        progress = (
            (iteration - 1)
            /
            (PSO_ITERATIONS - 1)
        )

    else:

        progress = 1.0


    inertia_weight = (

        PSO_W_START

        -

        (
            PSO_W_START
            -
            PSO_W_END
        )

        * progress
    )


    # =========================================================================
    # SAVE HISTORY
    # =========================================================================

    optimization_history.append(
        {

            "Iteration":
                iteration,

            "Iteration_Best_Fitness":
                iteration_best,

            "Iteration_Average_Fitness":
                iteration_average,

            "Iteration_Worst_Fitness":
                iteration_worst,

            "Global_Best_Fitness":
                global_best_score,

            "Inertia_Weight":
                inertia_weight,

            "Best_n_estimators":
                global_parameters[
                    "n_estimators"
                ],

            "Best_max_depth":
                global_parameters[
                    "max_depth"
                ],

            "Best_min_samples_split":
                global_parameters[
                    "min_samples_split"
                ],

            "Best_min_samples_leaf":
                global_parameters[
                    "min_samples_leaf"
                ],

            "Best_max_features":
                str(
                    global_parameters[
                        "max_features"
                    ]
                )
        }
    )


    print(
        "\nIteration best:",
        f"{iteration_best:.4f}"
    )


    print(
        "Iteration average:",
        f"{iteration_average:.4f}"
    )


    print(
        "Global best:",
        f"{global_best_score:.4f}"
    )


    print(
        "Global best parameters:"
    )


    print(
        global_parameters
    )


    # =========================================================================
    # 51. UPDATE VELOCITIES AND POSITIONS
    # =========================================================================

    for particle_index in range(
        PSO_SWARM_SIZE
    ):

        r1 = np.random.random(
            dimensions
        )


        r2 = np.random.random(
            dimensions
        )


        cognitive_component = (

            PSO_C1

            *

            r1

            *

            (
                personal_best_positions[
                    particle_index
                ]

                -

                positions[
                    particle_index
                ]
            )
        )


        social_component = (

            PSO_C2

            *

            r2

            *

            (
                global_best_position

                -

                positions[
                    particle_index
                ]
            )
        )


        velocities[
            particle_index
        ] = (

            inertia_weight

            *

            velocities[
                particle_index
            ]

            +

            cognitive_component

            +

            social_component
        )


        # =====================================================================
        # VELOCITY CLIPPING
        # =====================================================================

        maximum_velocity = (
            0.25
            *
            velocity_range
        )


        velocities[
            particle_index
        ] = np.clip(

            velocities[
                particle_index
            ],

            -maximum_velocity,

            maximum_velocity
        )


        # =====================================================================
        # POSITION UPDATE
        # =====================================================================

        positions[
            particle_index
        ] = (

            positions[
                particle_index
            ]

            +

            velocities[
                particle_index
            ]
        )


        # =====================================================================
        # POSITION BOUNDS
        # =====================================================================

        positions[
            particle_index
        ] = np.clip(

            positions[
                particle_index
            ],

            LOWER_BOUNDS,

            UPPER_BOUNDS
        )


# =============================================================================
# 52. PSO COMPLETE
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "PSO OPTIMIZATION COMPLETE"
)

print(
    "=" * 90
)


best_parameters = (
    position_to_parameters(
        global_best_position
    )
)


print(
    "\nBest PSO fitness:"
)

print(
    global_best_score
)


print(
    "\nBest PSO hyperparameters:"
)


for parameter, value in (
    best_parameters.items()
):

    print(
        f"{parameter}: {value}"
    )


# =============================================================================
# 53. SAVE OPTIMIZATION HISTORY
# =============================================================================

optimization_df = pd.DataFrame(
    optimization_history
)


optimization_df.to_csv(

    PSO_OPTIMIZATION_HISTORY_PATH,

    index=False
)


print(
    "\nOptimization history saved:"
)

print(
    PSO_OPTIMIZATION_HISTORY_PATH
)


# =============================================================================
# 54. PSO OPTIMIZATION GRAPH
# =============================================================================

plt.figure(
    figsize=(11, 7)
)


plt.plot(

    optimization_df[
        "Iteration"
    ],

    optimization_df[
        "Iteration_Best_Fitness"
    ],

    marker="o",

    label="Iteration Best"
)


plt.plot(

    optimization_df[
        "Iteration"
    ],

    optimization_df[
        "Iteration_Average_Fitness"
    ],

    marker="s",

    label="Iteration Average"
)


plt.plot(

    optimization_df[
        "Iteration"
    ],

    optimization_df[
        "Global_Best_Fitness"
    ],

    marker="^",

    linestyle="--",

    label="Global Best"
)


plt.xlabel(
    "PSO Iteration"
)


plt.ylabel(
    "Cross-Validation Accuracy"
)


plt.title(
    "PSO Optimization Progress"
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_OPTIMIZATION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    PSO_OPTIMIZATION_GRAPH_PATH
)


# =============================================================================
# 55. TRAIN FINAL PSO-OPTIMIZED MODEL
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "TRAINING FINAL PSO-OPTIMIZED RANDOM FOREST"
)

print(
    "=" * 90
)


pso_model = (
    create_pipeline(
        best_parameters
    )
)


pso_model.fit(
    X_train,
    y_train
)


# =============================================================================
# 56. PREDICT TEST DATA
# =============================================================================

pso_predictions = (
    pso_model.predict(
        X_test
    )
)


pso_probabilities = (
    pso_model.predict_proba(
        X_test
    )
)


# =============================================================================
# 57. PSO METRICS
# =============================================================================

pso_accuracy = accuracy_score(
    y_test,
    pso_predictions
)


pso_precision = precision_score(
    y_test,
    pso_predictions,
    average="weighted",
    zero_division=0
)


pso_recall = recall_score(
    y_test,
    pso_predictions,
    average="weighted",
    zero_division=0
)


pso_f1 = f1_score(
    y_test,
    pso_predictions,
    average="weighted",
    zero_division=0
)


print(
    "\n" + "=" * 90
)

print(
    "PSO MODEL PERFORMANCE"
)

print(
    "=" * 90
)


print(
    "\nAccuracy:",
    f"{pso_accuracy * 100:.2f}%"
)


print(
    "Precision:",
    f"{pso_precision * 100:.2f}%"
)


print(
    "Recall:",
    f"{pso_recall * 100:.2f}%"
)


print(
    "F1 Score:",
    f"{pso_f1 * 100:.2f}%"
)


print(
    "\nClassification Report:"
)


print(

    classification_report(

        y_test,

        pso_predictions,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Low",
            "Medium",
            "High"
        ],

        zero_division=0
    )

)


# =============================================================================
# 58. SAVE PSO MODEL
# =============================================================================

with open(
    PSO_MODEL_PATH,
    "wb"
) as file:

    pickle.dump(
        pso_model,
        file
    )


print(
    "\nPSO model saved:"
)

print(
    PSO_MODEL_PATH
)


# =============================================================================
# 59. SAVE PROCESSED DATA
# =============================================================================

working_df.to_csv(

    PSO_PROCESSED_PATH,

    index=False
)


print(
    "\nProcessed data saved:"
)

print(
    PSO_PROCESSED_PATH
)


# =============================================================================
# 60. CREATE PREDICTION CSV
# =============================================================================

prediction_df = (

    X_test

    .reset_index(
        drop=True
    )

    .copy()
)


actual_array = np.asarray(
    y_test
)


prediction_array = np.asarray(
    pso_predictions
)


prediction_df[
    "Actual_Class_Code"
] = actual_array


prediction_df[
    "Actual_Class"
] = [

    reverse_mapping[
        int(value)
    ]

    for value
    in actual_array
]


prediction_df[
    "PSO_Prediction_Code"
] = prediction_array


prediction_df[
    "PSO_Prediction"
] = [

    reverse_mapping[
        int(value)
    ]

    for value
    in prediction_array
]


prediction_df[
    "PSO_Correct"
] = (

    actual_array
    ==
    prediction_array
)


# =============================================================================
# 61. ADD PSO PROBABILITIES
# =============================================================================

probability_classes = (

    pso_model

    .named_steps[
        "classifier"
    ]

    .classes_
)


for probability_index, class_id in enumerate(
    probability_classes
):

    class_name = (
        reverse_mapping[
            int(class_id)
        ]
    )


    prediction_df[
        f"PSO_Probability_{class_name}"
    ] = (

        pso_probabilities[
            :,
            probability_index
        ]
    )


# =============================================================================
# 62. SAVE PREDICTION CSV
# =============================================================================

prediction_df.to_csv(

    PSO_PREDICTION_PATH,

    index=False
)


print(
    "\nPSO prediction CSV saved:"
)

print(
    PSO_PREDICTION_PATH
)


print(
    "\nPrediction Preview:"
)


print(
    prediction_df.head(10)
)


# =============================================================================
# 63. CREATE RESULT CSV
# =============================================================================

result_df = pd.DataFrame(
    {

        "Method": [
            "PSO-Optimized Random Forest"
        ],

        "PSO_CV_Best_Accuracy": [
            global_best_score
        ],

        "Test_Accuracy": [
            pso_accuracy
        ],

        "Test_Accuracy_Percentage": [
            pso_accuracy * 100
        ],

        "Precision": [
            pso_precision
        ],

        "Precision_Percentage": [
            pso_precision * 100
        ],

        "Recall": [
            pso_recall
        ],

        "Recall_Percentage": [
            pso_recall * 100
        ],

        "F1_Score": [
            pso_f1
        ],

        "F1_Percentage": [
            pso_f1 * 100
        ],

        "n_estimators": [
            best_parameters[
                "n_estimators"
            ]
        ],

        "max_depth": [
            best_parameters[
                "max_depth"
            ]
        ],

        "min_samples_split": [
            best_parameters[
                "min_samples_split"
            ]
        ],

        "min_samples_leaf": [
            best_parameters[
                "min_samples_leaf"
            ]
        ],

        "max_features": [
            str(
                best_parameters[
                    "max_features"
                ]
            )
        ]
    }
)


result_df.to_csv(

    PSO_RESULT_PATH,

    index=False
)


print(
    "\nPSO result CSV saved:"
)

print(
    PSO_RESULT_PATH
)


print(
    "\nPSO Result:"
)


print(
    result_df
)


# =============================================================================
# 64. PSO ACCURACY GRAPH
# =============================================================================

accuracy_labels = [

    "PSO CV Accuracy",

    "PSO Test Accuracy"
]


accuracy_values = [

    global_best_score * 100,

    pso_accuracy * 100
]


plt.figure(
    figsize=(8, 6)
)


bars = plt.bar(

    accuracy_labels,

    accuracy_values
)


plt.ylabel(
    "Accuracy (%)"
)


plt.title(
    "PSO Model Accuracy"
)


plt.ylim(
    0,
    100
)


for bar, value in zip(
    bars,
    accuracy_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center",

        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    PSO_ACCURACY_GRAPH_PATH
)


# =============================================================================
# 65. PSO PREDICTION GRAPH
# =============================================================================

number_of_samples = min(
    50,
    len(actual_array)
)


sample_numbers = np.arange(
    1,
    number_of_samples + 1
)


plt.figure(
    figsize=(14, 7)
)


plt.plot(

    sample_numbers,

    actual_array[
        :number_of_samples
    ],

    marker="o",

    linewidth=2,

    label="Actual"
)


plt.plot(

    sample_numbers,

    prediction_array[
        :number_of_samples
    ],

    marker="s",

    linestyle="--",

    linewidth=2,

    label="PSO Prediction"
)


plt.xlabel(
    "Test Sample"
)


plt.ylabel(
    "Production Performance"
)


plt.title(
    "Actual vs PSO Predicted Coal Mine Performance"
)


plt.yticks(

    [
        0,
        1,
        2
    ],

    [
        "Low",
        "Medium",
        "High"
    ]
)


plt.legend()


plt.grid(
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    PSO_PREDICTION_GRAPH_PATH
)


# =============================================================================
# 66. PSO RESULT GRAPH
# =============================================================================

metrics = [

    "Accuracy",

    "Precision",

    "Recall",

    "F1 Score"
]


pso_scores = [

    pso_accuracy * 100,

    pso_precision * 100,

    pso_recall * 100,

    pso_f1 * 100
]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(
    metrics,
    pso_scores
)


plt.xlabel(
    "Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "PSO-Optimized Model Performance"
)


plt.ylim(
    0,
    100
)


for bar, score in zip(
    bars,
    pso_scores
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        score + 1,

        f"{score:.2f}%",

        ha="center",

        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)


plt.tight_layout()


plt.savefig(

    PSO_RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    PSO_RESULT_GRAPH_PATH
)


# =============================================================================
# 67. TRAIN BASELINE RANDOM FOREST
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "TRAINING BASELINE RANDOM FOREST FOR COMPARISON"
)

print(
    "=" * 90
)


baseline_parameters = {

    "n_estimators": 300,

    "max_depth": 12,

    "min_samples_split": 2,

    "min_samples_leaf": 1,

    "max_features": "sqrt"
}


baseline_model = (
    create_pipeline(
        baseline_parameters
    )
)


baseline_model.fit(
    X_train,
    y_train
)


baseline_predictions = (
    baseline_model.predict(
        X_test
    )
)


# =============================================================================
# 68. BASELINE METRICS
# =============================================================================

baseline_accuracy = accuracy_score(
    y_test,
    baseline_predictions
)


baseline_precision = precision_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)


baseline_recall = recall_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)


baseline_f1 = f1_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)


print(
    "\nBaseline Accuracy:",
    f"{baseline_accuracy * 100:.2f}%"
)


print(
    "PSO Accuracy:",
    f"{pso_accuracy * 100:.2f}%"
)


# =============================================================================
# 69. PSO COMPARISON GRAPH
# =============================================================================

baseline_scores = [

    baseline_accuracy * 100,

    baseline_precision * 100,

    baseline_recall * 100,

    baseline_f1 * 100
]


pso_scores = [

    pso_accuracy * 100,

    pso_precision * 100,

    pso_recall * 100,

    pso_f1 * 100
]


x = np.arange(
    len(metrics)
)


width = 0.35


plt.figure(
    figsize=(11, 7)
)


baseline_bars = plt.bar(

    x - width / 2,

    baseline_scores,

    width,

    label="Baseline Random Forest"
)


pso_bars = plt.bar(

    x + width / 2,

    pso_scores,

    width,

    label="PSO-Optimized Random Forest"
)


plt.xlabel(
    "Evaluation Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "Baseline Random Forest vs PSO-Optimized Random Forest"
)


plt.xticks(
    x,
    metrics
)


plt.ylim(
    0,
    100
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


# =============================================================================
# 70. VALUES ON COMPARISON BARS
# =============================================================================

for bars in [

    baseline_bars,

    pso_bars

]:

    for bar in bars:

        height = (
            bar.get_height()
        )


        plt.text(

            bar.get_x()
            +
            bar.get_width() / 2,

            height + 0.5,

            f"{height:.1f}",

            ha="center",

            fontsize=8
        )


plt.tight_layout()


plt.savefig(

    PSO_COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    PSO_COMPARISON_GRAPH_PATH
)


# =============================================================================
# 71. PSO CONFUSION MATRIX
# =============================================================================

pso_cm = confusion_matrix(

    y_test,

    pso_predictions,

    labels=[
        0,
        1,
        2
    ]
)


print(
    "\nPSO Confusion Matrix:"
)


print(
    pso_cm
)


# =============================================================================
# 72. PSO HEATMAP
# =============================================================================

class_names = [

    "Low",

    "Medium",

    "High"
]


plt.figure(
    figsize=(8, 7)
)


image = plt.imshow(

    pso_cm,

    interpolation="nearest",

    cmap="Blues"
)


plt.colorbar(
    image
)


plt.title(
    "PSO-Optimized Model Confusion Matrix"
)


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "Actual Class"
)


tick_positions = np.arange(
    len(class_names)
)


plt.xticks(
    tick_positions,
    class_names
)


plt.yticks(
    tick_positions,
    class_names
)


# =============================================================================
# 73. ADD VALUES TO HEATMAP
# =============================================================================

threshold = (

    pso_cm.max() / 2

    if pso_cm.size > 0

    else 0
)


for row in range(
    pso_cm.shape[0]
):

    for column in range(
        pso_cm.shape[1]
    ):

        value = (
            pso_cm[
                row,
                column
            ]
        )


        if value > threshold:

            text_color = (
                "white"
            )

        else:

            text_color = (
                "black"
            )


        plt.text(

            column,

            row,

            str(value),

            horizontalalignment="center",

            verticalalignment="center",

            color=text_color,

            fontsize=14,

            fontweight="bold"
        )


plt.tight_layout()


plt.savefig(

    PSO_HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    PSO_HEATMAP_PATH
)


# =============================================================================
# 74. CLASS DISTRIBUTION GRAPH
# =============================================================================

actual_counts = [

    np.sum(
        actual_array == class_id
    )

    for class_id in [
        0,
        1,
        2
    ]
]


predicted_counts = [

    np.sum(
        prediction_array == class_id
    )

    for class_id in [
        0,
        1,
        2
    ]
]


x = np.arange(
    len(class_names)
)


width = 0.35


plt.figure(
    figsize=(10, 6)
)


actual_bars = plt.bar(

    x - width / 2,

    actual_counts,

    width,

    label="Actual"
)


predicted_bars = plt.bar(

    x + width / 2,

    predicted_counts,

    width,

    label="PSO Predicted"
)


plt.xlabel(
    "Production Performance Class"
)


plt.ylabel(
    "Number of Samples"
)


plt.title(
    "Actual vs PSO Predicted Class Distribution"
)


plt.xticks(
    x,
    class_names
)


plt.legend()


plt.grid(
    axis="y",
    alpha=0.3
)


# =============================================================================
# 75. VALUES ON DISTRIBUTION BARS
# =============================================================================

for bars in [

    actual_bars,

    predicted_bars

]:

    for bar in bars:

        height = (
            bar.get_height()
        )


        plt.text(

            bar.get_x()
            +
            bar.get_width() / 2,

            height + 0.2,

            f"{int(height)}",

            ha="center",

            fontsize=9
        )


plt.tight_layout()


plt.savefig(

    PSO_CLASS_DISTRIBUTION_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


print(
    "\nSaved:"
)

print(
    PSO_CLASS_DISTRIBUTION_PATH
)


# =============================================================================
# 76. SAVE YAML CONFIGURATION
# =============================================================================

yaml_config = {

    "project": {

        "name":
            "AI-Based Coal Mine Production Analysis "
            "and Performance Classification",

        "optimization":
            "Particle Swarm Optimization",

        "classifier":
            "Random Forest"
    },


    "dataset": {

        "path":
            CSV_PATH,

        "mine_column":
            mine_column,

        "company_column":
            company_column,

        "production_column":
            production_column,

        "features":
            feature_columns,

        "valid_records":
            int(
                len(
                    working_df
                )
            )
    },


    "target": {

        "classes": {

            0: "Low",

            1: "Medium",

            2: "High"
        },

        "low_medium_threshold":
            float(
                q33
            ),

        "medium_high_threshold":
            float(
                q66
            )
    },


    "pso": {

        "swarm_size":
            PSO_SWARM_SIZE,

        "iterations":
            PSO_ITERATIONS,

        "inertia_start":
            PSO_W_START,

        "inertia_end":
            PSO_W_END,

        "cognitive_coefficient":
            PSO_C1,

        "social_coefficient":
            PSO_C2,

        "best_cv_fitness":
            float(
                global_best_score
            ),

        "best_parameters": {

            "n_estimators":
                int(
                    best_parameters[
                        "n_estimators"
                    ]
                ),

            "max_depth":
                int(
                    best_parameters[
                        "max_depth"
                    ]
                ),

            "min_samples_split":
                int(
                    best_parameters[
                        "min_samples_split"
                    ]
                ),

            "min_samples_leaf":
                int(
                    best_parameters[
                        "min_samples_leaf"
                    ]
                ),

            "max_features":
                str(
                    best_parameters[
                        "max_features"
                    ]
                )
        }
    },


    "performance": {

        "accuracy":
            float(
                pso_accuracy
            ),

        "precision":
            float(
                pso_precision
            ),

        "recall":
            float(
                pso_recall
            ),

        "f1_score":
            float(
                pso_f1
            )
    }
}


with open(

    PSO_CONFIG_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_config,

        file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "\nPSO YAML saved:"
)

print(
    PSO_CONFIG_PATH
)


# =============================================================================
# 77. JSON SERIALIZATION HELPER
# =============================================================================

def serializable(
    value
):

    if isinstance(
        value,
        np.integer
    ):

        return int(
            value
        )


    if isinstance(
        value,
        np.floating
    ):

        return float(
            value
        )


    if isinstance(
        value,
        np.ndarray
    ):

        return value.tolist()


    if isinstance(
        value,
        dict
    ):

        return {

            str(key):
                serializable(
                    item
                )

            for key, item
            in value.items()
        }


    if isinstance(
        value,
        list
    ):

        return [

            serializable(
                item
            )

            for item
            in value
        ]


    return value


# =============================================================================
# 78. CLASSIFICATION REPORT DICTIONARY
# =============================================================================

pso_report = classification_report(

    y_test,

    pso_predictions,

    labels=[
        0,
        1,
        2
    ],

    target_names=[
        "Low",
        "Medium",
        "High"
    ],

    zero_division=0,

    output_dict=True
)


# =============================================================================
# 79. SAVE JSON METADATA
# =============================================================================

metadata = {

    "project":
        "AI-Based Coal Mine Production Analysis "
        "and Performance Classification",

    "optimization":
        "Particle Swarm Optimization",

    "model":
        "PSO-Optimized Random Forest",

    "records":
        int(
            len(
                working_df
            )
        ),

    "training_records":
        int(
            len(
                X_train
            )
        ),

    "testing_records":
        int(
            len(
                X_test
            )
        ),

    "features":
        feature_columns,

    "target":
        TARGET_COLUMN,

    "production_thresholds": {

        "low_medium":
            float(
                q33
            ),

        "medium_high":
            float(
                q66
            )
    },

    "pso_settings": {

        "swarm_size":
            PSO_SWARM_SIZE,

        "iterations":
            PSO_ITERATIONS,

        "w_start":
            PSO_W_START,

        "w_end":
            PSO_W_END,

        "c1":
            PSO_C1,

        "c2":
            PSO_C2
    },

    "best_parameters":
        best_parameters,

    "best_cross_validation_accuracy":
        float(
            global_best_score
        ),

    "test_metrics": {

        "accuracy":
            float(
                pso_accuracy
            ),

        "precision":
            float(
                pso_precision
            ),

        "recall":
            float(
                pso_recall
            ),

        "f1_score":
            float(
                pso_f1
            )
    },

    "baseline_metrics": {

        "accuracy":
            float(
                baseline_accuracy
            ),

        "precision":
            float(
                baseline_precision
            ),

        "recall":
            float(
                baseline_recall
            ),

        "f1_score":
            float(
                baseline_f1
            )
    },

    "confusion_matrix":
        pso_cm,

    "classification_report":
        pso_report
}


with open(

    PSO_METADATA_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        serializable(
            metadata
        ),

        file,

        indent=4,

        ensure_ascii=False
    )


print(
    "\nPSO JSON saved:"
)

print(
    PSO_METADATA_PATH
)


# =============================================================================
# 80. VERIFY GENERATED FILES
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "VERIFYING PSO OUTPUT FILES"
)

print(
    "=" * 90
)


generated_files = [

    PSO_MODEL_PATH,

    PSO_CONFIG_PATH,

    PSO_METADATA_PATH,

    PSO_PROCESSED_PATH,

    PSO_PREDICTION_PATH,

    PSO_RESULT_PATH,

    PSO_OPTIMIZATION_HISTORY_PATH,

    PSO_ACCURACY_GRAPH_PATH,

    PSO_PREDICTION_GRAPH_PATH,

    PSO_RESULT_GRAPH_PATH,

    PSO_COMPARISON_GRAPH_PATH,

    PSO_HEATMAP_PATH,

    PSO_OPTIMIZATION_GRAPH_PATH,

    PSO_CLASS_DISTRIBUTION_PATH
]


for path in generated_files:

    if os.path.exists(
        path
    ):

        size = os.path.getsize(
            path
        )


        print(

            f"[OK] "
            f"{os.path.basename(path)} "
            f"({size:,} bytes)"
        )

    else:

        print(

            f"[MISSING] "
            f"{os.path.basename(path)}"
        )


# =============================================================================
# 81. IMPROVEMENT CALCULATION
# =============================================================================

accuracy_improvement = (

    (
        pso_accuracy
        -
        baseline_accuracy
    )

    * 100
)


f1_improvement = (

    (
        pso_f1
        -
        baseline_f1
    )

    * 100
)


# =============================================================================
# 82. FINAL SUMMARY
# =============================================================================

print(
    "\n" + "=" * 90
)

print(
    "PSO PROJECT COMPLETED SUCCESSFULLY"
)

print(
    "=" * 90
)


print(
    f"""

PROJECT
=======

AI-Based Coal Mine Production Analysis
and Performance Classification


OPTIMIZATION
============

Particle Swarm Optimization (PSO)


CLASSIFIER
==========

Random Forest Classifier


PSO SETTINGS
============

Swarm Size:
{PSO_SWARM_SIZE}

Iterations:
{PSO_ITERATIONS}

Initial Inertia:
{PSO_W_START}

Final Inertia:
{PSO_W_END}

Cognitive Coefficient (C1):
{PSO_C1}

Social Coefficient (C2):
{PSO_C2}


BEST PSO PARAMETERS
===================

n_estimators:
{best_parameters['n_estimators']}

max_depth:
{best_parameters['max_depth']}

min_samples_split:
{best_parameters['min_samples_split']}

min_samples_leaf:
{best_parameters['min_samples_leaf']}

max_features:
{best_parameters['max_features']}


BEST CROSS-VALIDATION FITNESS
=============================

{global_best_score * 100:.2f}%


BASELINE RANDOM FOREST
======================

Accuracy:
{baseline_accuracy * 100:.2f}%

Precision:
{baseline_precision * 100:.2f}%

Recall:
{baseline_recall * 100:.2f}%

F1 Score:
{baseline_f1 * 100:.2f}%


PSO-OPTIMIZED RANDOM FOREST
===========================

Accuracy:
{pso_accuracy * 100:.2f}%

Precision:
{pso_precision * 100:.2f}%

Recall:
{pso_recall * 100:.2f}%

F1 Score:
{pso_f1 * 100:.2f}%


CHANGE FROM BASELINE
====================

Accuracy change:
{accuracy_improvement:+.2f} percentage points

F1 Score change:
{f1_improvement:+.2f} percentage points


GENERATED FILES
===============

1. pso_model.pkl

2. pso_config.yaml

3. pso_metadata.json

4. pso_processed_data.csv

5. pso_prediction.csv

6. pso_result.csv

7. pso_optimization_history.csv

8. pso_accuracy_graph.png

9. pso_prediction_graph.png

10. pso_result_graph.png

11. pso_comparison_graph.png

12. pso_heatmap.png

13. pso_optimization_graph.png

14. pso_class_distribution_graph.png


OUTPUT DIRECTORY
================

{PROJECT_DIR}

"""
)


print(
    "=" * 90
)

print(
    "ALL PSO RESULTS SAVED WITH pso_ PREFIX"
)

print(
    "=" * 90
)
```
