```python
import os
import re
import json
import yaml
import pickle
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score
)

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)


# =============================================================================
# 2. SETTINGS
# =============================================================================

warnings.filterwarnings("ignore")

RANDOM_STATE = 42

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)


# =============================================================================
# 3. PROJECT DIRECTORY
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Coal Mine Production Analysis and Performance Classification"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "rs_session-241_au482_1.1.csv"
)

os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 4. OUTPUT PATHS
# =============================================================================

AIS_MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "ais_model.pkl"
)

AIS_CONFIG_PATH = os.path.join(
    PROJECT_DIR,
    "ais_config.yaml"
)

AIS_METADATA_PATH = os.path.join(
    PROJECT_DIR,
    "ais_metadata.json"
)

AIS_PROCESSED_PATH = os.path.join(
    PROJECT_DIR,
    "ais_processed_data.csv"
)

AIS_PREDICTION_PATH = os.path.join(
    PROJECT_DIR,
    "ais_prediction.csv"
)

AIS_RESULT_PATH = os.path.join(
    PROJECT_DIR,
    "ais_result.csv"
)

AIS_OPTIMIZATION_HISTORY_PATH = os.path.join(
    PROJECT_DIR,
    "ais_optimization_history.csv"
)


# Graphs

AIS_ACCURACY_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_accuracy_graph.png"
)

AIS_PREDICTION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_prediction_graph.png"
)

AIS_RESULT_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_result_graph.png"
)

AIS_COMPARISON_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_comparison_graph.png"
)

AIS_HEATMAP_PATH = os.path.join(
    PROJECT_DIR,
    "ais_heatmap.png"
)

AIS_OPTIMIZATION_GRAPH_PATH = os.path.join(
    PROJECT_DIR,
    "ais_optimization_graph.png"
)

AIS_CLASS_DISTRIBUTION_PATH = os.path.join(
    PROJECT_DIR,
    "ais_class_distribution_graph.png"
)


# =============================================================================
# 5. HEADER
# =============================================================================

print("=" * 90)
print("AI-BASED COAL MINE PRODUCTION ANALYSIS")
print("ARTIFICIAL IMMUNE SYSTEM (AIS) OPTIMIZATION")
print("=" * 90)


# =============================================================================
# 6. CHECK DATASET
# =============================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}"
    )


print("\nDataset found:")
print(CSV_PATH)


# =============================================================================
# 7. LOAD DATASET
# =============================================================================

print("\n" + "=" * 90)
print("LOADING DATASET")
print("=" * 90)


encodings = [
    "utf-8",
    "utf-8-sig",
    "cp1252",
    "latin1"
]


df = None
used_encoding = None


for encoding in encodings:

    try:

        df = pd.read_csv(
            CSV_PATH,
            encoding=encoding
        )

        used_encoding = encoding

        break

    except UnicodeDecodeError:

        continue


if df is None:

    raise ValueError(
        "Could not read CSV using common encodings."
    )


print("\nDataset loaded successfully.")

print("Encoding:", used_encoding)
print("Shape:", df.shape)


# =============================================================================
# 8. CLEAN COLUMN NAMES
# =============================================================================

def clean_column_name(column):

    column = str(column)

    column = column.replace(
        "\n",
        " "
    )

    column = column.replace(
        "\r",
        " "
    )

    column = re.sub(
        r"\s+",
        " ",
        column
    )

    return column.strip()


df.columns = [
    clean_column_name(column)
    for column in df.columns
]


# =============================================================================
# 9. MAKE COLUMN NAMES UNIQUE
# =============================================================================

def make_unique_columns(columns):

    counts = {}

    output = []

    for column in columns:

        if column not in counts:

            counts[column] = 0

            output.append(
                column
            )

        else:

            counts[column] += 1

            new_name = (
                f"{column}_{counts[column]}"
            )

            while new_name in output:

                counts[column] += 1

                new_name = (
                    f"{column}_{counts[column]}"
                )

            output.append(
                new_name
            )

    return output


df.columns = make_unique_columns(
    df.columns
)


print("\nColumns:")

for index, column in enumerate(
    df.columns,
    start=1
):

    print(
        f"{index}. {column}"
    )


# =============================================================================
# 10. REMOVE EMPTY ROWS/COLUMNS
# =============================================================================

df = df.dropna(
    axis=0,
    how="all"
)

df = df.dropna(
    axis=1,
    how="all"
)


# =============================================================================
# 11. COLUMN SEARCH
# =============================================================================

def find_column(
    columns,
    keywords,
    exclude=None
):

    if exclude is None:
        exclude = []

    # Exact match

    for keyword in keywords:

        for column in columns:

            if column in exclude:
                continue

            if (
                str(column).lower().strip()
                ==
                keyword.lower().strip()
            ):

                return column


    # Partial match

    for keyword in keywords:

        for column in columns:

            if column in exclude:
                continue

            if (
                keyword.lower()
                in str(column).lower()
            ):

                return column

    return None


# =============================================================================
# 12. DETECT PRODUCTION COLUMN
# =============================================================================

production_column = find_column(

    df.columns,

    [
        "quantity of coal production",
        "coal production quantity",
        "coal production",
        "production quantity",
        "production",
        "quantity"
    ]
)


# =============================================================================
# 13. NUMERIC FALLBACK
# =============================================================================

if production_column is None:

    best_column = None
    best_ratio = -1

    for column in df.columns:

        values = (
            df[column]
            .astype(str)
            .str.replace(
                ",",
                "",
                regex=False
            )
            .str.strip()
        )

        numeric_values = pd.to_numeric(
            values,
            errors="coerce"
        )

        ratio = (
            numeric_values
            .notna()
            .mean()
        )

        if ratio > best_ratio:

            best_ratio = ratio
            best_column = column

    production_column = best_column


# =============================================================================
# 14. DETECT MINE
# =============================================================================

mine_column = find_column(

    df.columns,

    [
        "name of mine",
        "mine name",
        "mine"
    ],

    exclude=[
        production_column
    ]
)


# =============================================================================
# 15. DETECT COMPANY
# =============================================================================

company_column = find_column(

    df.columns,

    [
        "coal company",
        "company name",
        "name of company",
        "company",
        "subsidiary",
        "region"
    ],

    exclude=[
        production_column,
        mine_column
    ]
)


# =============================================================================
# 16. FALLBACK CATEGORICAL COLUMNS
# =============================================================================

available_columns = [

    column

    for column in df.columns

    if column != production_column
]


if mine_column is None:

    if len(available_columns) >= 1:

        mine_column = available_columns[0]


if company_column is None:

    remaining = [

        column

        for column in available_columns

        if column != mine_column
    ]

    if len(remaining) >= 1:

        company_column = remaining[0]


# =============================================================================
# 17. DISPLAY COLUMNS
# =============================================================================

print("\n" + "=" * 90)
print("DETECTED COLUMNS")
print("=" * 90)

print("\nMine:")
print(mine_column)

print("\nCompany:")
print(company_column)

print("\nProduction:")
print(production_column)


if production_column is None:

    raise ValueError(
        "Production column could not be identified."
    )


# =============================================================================
# 18. WORKING DATA
# =============================================================================

working_df = df.copy()


# =============================================================================
# 19. CLEAN CATEGORICAL VALUES
# =============================================================================

def clean_categorical_series(series):

    series = series.copy()

    series = series.fillna(
        "Unknown"
    )

    series = series.astype(
        str
    )

    series = series.map(
        lambda value:
        value.strip()
    )

    series = series.replace(

        [
            "",
            "nan",
            "NaN",
            "None",
            "none",
            "NULL",
            "null"
        ],

        "Unknown"
    )

    return series


if mine_column is not None:

    working_df[
        mine_column
    ] = clean_categorical_series(

        working_df[
            mine_column
        ]
    )


if company_column is not None:

    working_df[
        company_column
    ] = clean_categorical_series(

        working_df[
            company_column
        ]
    )


# =============================================================================
# 20. CLEAN PRODUCTION
# =============================================================================

production_values = (

    working_df[
        production_column
    ]

    .astype(str)

    .str.replace(
        ",",
        "",
        regex=False
    )

    .str.replace(
        "₹",
        "",
        regex=False
    )

    .str.replace(
        r"[^0-9.\-]",
        "",
        regex=True
    )

    .str.strip()
)


working_df[
    production_column
] = pd.to_numeric(

    production_values,

    errors="coerce"
)


# =============================================================================
# 21. REMOVE INVALID PRODUCTION
# =============================================================================

working_df = (

    working_df

    .dropna(
        subset=[
            production_column
        ]
    )

    .copy()
)


working_df = (

    working_df[

        working_df[
            production_column
        ] >= 0

    ]

    .copy()
)


working_df.reset_index(
    drop=True,
    inplace=True
)


if len(working_df) < 10:

    raise ValueError(
        "Too few valid records."
    )


print(
    "\nValid records:",
    len(working_df)
)


# =============================================================================
# 22. PRODUCTION STATISTICS
# =============================================================================

print("\n" + "=" * 90)
print("PRODUCTION STATISTICS")
print("=" * 90)


print(

    working_df[
        production_column
    ].describe()

)


# =============================================================================
# 23. CREATE TARGET
# =============================================================================

TARGET_COLUMN = (
    "Production_Performance"
)


q33 = float(

    working_df[
        production_column
    ].quantile(
        1 / 3
    )
)


q66 = float(

    working_df[
        production_column
    ].quantile(
        2 / 3
    )
)


print(
    "\nLow / Medium threshold:",
    q33
)

print(
    "Medium / High threshold:",
    q66
)


# =============================================================================
# 24. CREATE QUANTILE CLASSES
# =============================================================================

try:

    working_df[
        TARGET_COLUMN
    ] = pd.qcut(

        working_df[
            production_column
        ],

        q=3,

        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


except ValueError:

    ranked = (

        working_df[
            production_column
        ]

        .rank(
            method="first"
        )
    )


    working_df[
        TARGET_COLUMN
    ] = pd.qcut(

        ranked,

        q=3,

        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


# =============================================================================
# 25. TARGET MAPPING
# =============================================================================

class_mapping = {

    "Low": 0,

    "Medium": 1,

    "High": 2
}


reverse_mapping = {

    0: "Low",

    1: "Medium",

    2: "High"
}


working_df[
    "Target"
] = (

    working_df[
        TARGET_COLUMN
    ]

    .astype(str)

    .map(
        class_mapping
    )
)


print(
    "\nClass distribution:"
)


print(

    working_df[
        TARGET_COLUMN
    ].value_counts()

)


# =============================================================================
# 26. FEATURES
# =============================================================================

feature_columns = []


if mine_column is not None:

    feature_columns.append(
        mine_column
    )


if (
    company_column is not None
    and company_column not in feature_columns
):

    feature_columns.append(
        company_column
    )


# Production cannot be used because it created target

feature_columns = [

    column

    for column in feature_columns

    if column not in [

        production_column,

        TARGET_COLUMN,

        "Target"
    ]
]


feature_columns = list(
    dict.fromkeys(
        feature_columns
    )
)


if len(feature_columns) == 0:

    raise ValueError(
        "No valid features found."
    )


print(
    "\nFeatures:"
)

print(
    feature_columns
)


# =============================================================================
# 27. X AND Y
# =============================================================================

X = working_df[
    feature_columns
].copy()


y = working_df[
    "Target"
].copy()


# Remove duplicate feature columns

X = X.loc[
    :,
    ~X.columns.duplicated()
].copy()


feature_columns = (
    X.columns.tolist()
)


# =============================================================================
# 28. CLEAN FEATURES
# =============================================================================

for column_index in range(
    X.shape[1]
):

    X.iloc[
        :,
        column_index
    ] = clean_categorical_series(

        X.iloc[
            :,
            column_index
        ]
    )


# =============================================================================
# 29. TRAIN TEST SPLIT
# =============================================================================

X_train, X_test, y_train, y_test = (
    train_test_split(

        X,
        y,

        test_size=0.20,

        random_state=RANDOM_STATE,

        stratify=y
    )
)


print(
    "\nTraining:",
    len(X_train)
)

print(
    "Testing:",
    len(X_test)
)


# =============================================================================
# 30. ENCODER FUNCTION
# =============================================================================

def create_encoder():

    try:

        return OneHotEncoder(

            handle_unknown="ignore",

            sparse_output=False
        )

    except TypeError:

        return OneHotEncoder(

            handle_unknown="ignore",

            sparse=False
        )


# =============================================================================
# 31. AIS PARAMETER SPACE
# =============================================================================
#
# An antibody represents:
#
# [n_estimators,
#  max_depth,
#  min_samples_split,
#  min_samples_leaf,
#  max_features]
#
# =============================================================================

PARAMETER_BOUNDS = {

    "n_estimators": (
        100,
        500
    ),

    "max_depth": (
        3,
        25
    ),

    "min_samples_split": (
        2,
        15
    ),

    "min_samples_leaf": (
        1,
        10
    )
}


MAX_FEATURE_OPTIONS = [

    "sqrt",

    "log2",

    None
]


# =============================================================================
# 32. AIS SETTINGS
# =============================================================================

AIS_POPULATION_SIZE = 10

AIS_GENERATIONS = 10

AIS_CLONE_FACTOR = 4

AIS_ELITE_SIZE = 4

AIS_RANDOM_NEW = 2

AIS_MUTATION_RATE = 0.35


print("\n" + "=" * 90)
print("AIS CONFIGURATION")
print("=" * 90)

print(
    "Population:",
    AIS_POPULATION_SIZE
)

print(
    "Generations:",
    AIS_GENERATIONS
)

print(
    "Clone factor:",
    AIS_CLONE_FACTOR
)

print(
    "Elite antibodies:",
    AIS_ELITE_SIZE
)

print(
    "Mutation rate:",
    AIS_MUTATION_RATE
)


# =============================================================================
# 33. CREATE RANDOM ANTIBODY
# =============================================================================

def create_random_antibody():

    antibody = {

        "n_estimators":
            random.randint(
                PARAMETER_BOUNDS[
                    "n_estimators"
                ][0],

                PARAMETER_BOUNDS[
                    "n_estimators"
                ][1]
            ),

        "max_depth":
            random.randint(
                PARAMETER_BOUNDS[
                    "max_depth"
                ][0],

                PARAMETER_BOUNDS[
                    "max_depth"
                ][1]
            ),

        "min_samples_split":
            random.randint(
                PARAMETER_BOUNDS[
                    "min_samples_split"
                ][0],

                PARAMETER_BOUNDS[
                    "min_samples_split"
                ][1]
            ),

        "min_samples_leaf":
            random.randint(
                PARAMETER_BOUNDS[
                    "min_samples_leaf"
                ][0],

                PARAMETER_BOUNDS[
                    "min_samples_leaf"
                ][1]
            ),

        "max_features":
            random.choice(
                MAX_FEATURE_OPTIONS
            )
    }

    return antibody


# =============================================================================
# 34. CREATE MODEL FROM ANTIBODY
# =============================================================================

def create_pipeline_from_antibody(
    antibody
):

    encoder = (
        create_encoder()
    )


    preprocessor = ColumnTransformer(

        transformers=[

            (
                "categorical",

                encoder,

                feature_columns
            )

        ],

        remainder="drop"
    )


    classifier = RandomForestClassifier(

        n_estimators=int(
            antibody[
                "n_estimators"
            ]
        ),

        max_depth=int(
            antibody[
                "max_depth"
            ]
        ),

        min_samples_split=int(
            antibody[
                "min_samples_split"
            ]
        ),

        min_samples_leaf=int(
            antibody[
                "min_samples_leaf"
            ]
        ),

        max_features=antibody[
            "max_features"
        ],

        class_weight="balanced",

        random_state=RANDOM_STATE,

        n_jobs=-1
    )


    pipeline = Pipeline(

        steps=[

            (
                "preprocessor",
                preprocessor
            ),

            (
                "classifier",
                classifier
            )
        ]
    )


    return pipeline


# =============================================================================
# 35. FITNESS FUNCTION
# =============================================================================

cv = StratifiedKFold(

    n_splits=3,

    shuffle=True,

    random_state=RANDOM_STATE
)


fitness_cache = {}


def antibody_key(
    antibody
):

    return (

        antibody[
            "n_estimators"
        ],

        antibody[
            "max_depth"
        ],

        antibody[
            "min_samples_split"
        ],

        antibody[
            "min_samples_leaf"
        ],

        str(
            antibody[
                "max_features"
            ]
        )
    )


def evaluate_antibody(
    antibody
):

    key = antibody_key(
        antibody
    )


    if key in fitness_cache:

        return fitness_cache[
            key
        ]


    model = (
        create_pipeline_from_antibody(
            antibody
        )
    )


    scores = cross_val_score(

        model,

        X_train,

        y_train,

        cv=cv,

        scoring="accuracy",

        n_jobs=1
    )


    fitness = float(
        np.mean(
            scores
        )
    )


    fitness_cache[
        key
    ] = fitness


    return fitness


# =============================================================================
# 36. MUTATION
# =============================================================================

def mutate_antibody(
    antibody,
    mutation_strength=1.0
):

    mutated = (
        antibody.copy()
    )


    # -------------------------------------------------------------------------
    # n_estimators
    # -------------------------------------------------------------------------

    if random.random() < AIS_MUTATION_RATE:

        change = random.randint(
            -80,
            80
        )

        change = int(
            change
            * mutation_strength
        )

        mutated[
            "n_estimators"
        ] += change


    # -------------------------------------------------------------------------
    # max_depth
    # -------------------------------------------------------------------------

    if random.random() < AIS_MUTATION_RATE:

        change = random.randint(
            -5,
            5
        )

        mutated[
            "max_depth"
        ] += change


    # -------------------------------------------------------------------------
    # min_samples_split
    # -------------------------------------------------------------------------

    if random.random() < AIS_MUTATION_RATE:

        change = random.randint(
            -3,
            3
        )

        mutated[
            "min_samples_split"
        ] += change


    # -------------------------------------------------------------------------
    # min_samples_leaf
    # -------------------------------------------------------------------------

    if random.random() < AIS_MUTATION_RATE:

        change = random.randint(
            -2,
            2
        )

        mutated[
            "min_samples_leaf"
        ] += change


    # -------------------------------------------------------------------------
    # max_features
    # -------------------------------------------------------------------------

    if random.random() < AIS_MUTATION_RATE:

        mutated[
            "max_features"
        ] = random.choice(
            MAX_FEATURE_OPTIONS
        )


    # -------------------------------------------------------------------------
    # Enforce parameter bounds
    # -------------------------------------------------------------------------

    mutated[
        "n_estimators"
    ] = int(
        np.clip(

            mutated[
                "n_estimators"
            ],

            PARAMETER_BOUNDS[
                "n_estimators"
            ][0],

            PARAMETER_BOUNDS[
                "n_estimators"
            ][1]
        )
    )


    mutated[
        "max_depth"
    ] = int(
        np.clip(

            mutated[
                "max_depth"
            ],

            PARAMETER_BOUNDS[
                "max_depth"
            ][0],

            PARAMETER_BOUNDS[
                "max_depth"
            ][1]
        )
    )


    mutated[
        "min_samples_split"
    ] = int(
        np.clip(

            mutated[
                "min_samples_split"
            ],

            PARAMETER_BOUNDS[
                "min_samples_split"
            ][0],

            PARAMETER_BOUNDS[
                "min_samples_split"
            ][1]
        )
    )


    mutated[
        "min_samples_leaf"
    ] = int(
        np.clip(

            mutated[
                "min_samples_leaf"
            ],

            PARAMETER_BOUNDS[
                "min_samples_leaf"
            ][0],

            PARAMETER_BOUNDS[
                "min_samples_leaf"
            ][1]
        )
    )


    return mutated


# =============================================================================
# 37. INITIAL AIS POPULATION
# =============================================================================

print("\n" + "=" * 90)
print("CREATING INITIAL ANTIBODY POPULATION")
print("=" * 90)


population = [

    create_random_antibody()

    for _ in range(
        AIS_POPULATION_SIZE
    )
]


# =============================================================================
# 38. AIS OPTIMIZATION
# =============================================================================

print("\n" + "=" * 90)
print("STARTING AIS CLONAL SELECTION OPTIMIZATION")
print("=" * 90)


optimization_history = []

global_best_antibody = None

global_best_fitness = -1


for generation in range(
    1,
    AIS_GENERATIONS + 1
):

    print(
        "\nGeneration",
        generation,
        "/",
        AIS_GENERATIONS
    )


    # =========================================================================
    # Evaluate current population
    # =========================================================================

    evaluated_population = []


    for index, antibody in enumerate(
        population,
        start=1
    ):

        fitness = evaluate_antibody(
            antibody
        )


        evaluated_population.append(
            (
                antibody,
                fitness
            )
        )


        print(
            f"Antibody {index:02d} | "
            f"Fitness = {fitness:.4f}"
        )


    # =========================================================================
    # Sort highest fitness first
    # =========================================================================

    evaluated_population.sort(

        key=lambda item:
        item[1],

        reverse=True
    )


    generation_best_antibody = (
        evaluated_population[
            0
        ][0]
    )


    generation_best_fitness = (
        evaluated_population[
            0
        ][1]
    )


    generation_average_fitness = float(
        np.mean(
            [
                item[1]
                for item
                in evaluated_population
            ]
        )
    )


    # =========================================================================
    # Update global memory cell
    # =========================================================================

    if (
        generation_best_fitness
        >
        global_best_fitness
    ):

        global_best_fitness = (
            generation_best_fitness
        )

        global_best_antibody = (
            generation_best_antibody.copy()
        )


    # =========================================================================
    # Store generation results
    # =========================================================================

    optimization_history.append(
        {

            "Generation":
                generation,

            "Best_Fitness":
                generation_best_fitness,

            "Average_Fitness":
                generation_average_fitness,

            "Global_Best_Fitness":
                global_best_fitness,

            "Best_n_estimators":
                generation_best_antibody[
                    "n_estimators"
                ],

            "Best_max_depth":
                generation_best_antibody[
                    "max_depth"
                ],

            "Best_min_samples_split":
                generation_best_antibody[
                    "min_samples_split"
                ],

            "Best_min_samples_leaf":
                generation_best_antibody[
                    "min_samples_leaf"
                ],

            "Best_max_features":
                str(
                    generation_best_antibody[
                        "max_features"
                    ]
                )
        }
    )


    print(
        "\nGeneration Best Fitness:",
        f"{generation_best_fitness:.4f}"
    )


    print(
        "Global Best Fitness:",
        f"{global_best_fitness:.4f}"
    )


    print(
        "Best Antibody:",
        generation_best_antibody
    )


    # =========================================================================
    # Select elite antibodies
    # =========================================================================

    elites = (

        evaluated_population[
            :AIS_ELITE_SIZE
        ]
    )


    # =========================================================================
    # Clonal expansion + hypermutation
    # =========================================================================

    clone_population = []


    for rank, (
        antibody,
        fitness
    ) in enumerate(
        elites
    ):

        # Better antibodies mutate less

        mutation_strength = (
            0.5
            +
            (
                rank
                /
                max(
                    1,
                    AIS_ELITE_SIZE - 1
                )
            )
        )


        number_of_clones = max(

            1,

            AIS_CLONE_FACTOR
            * (
                AIS_ELITE_SIZE
                - rank
            )
        )


        for _ in range(
            number_of_clones
        ):

            clone = mutate_antibody(

                antibody,

                mutation_strength
            )

            clone_population.append(
                clone
            )


    # =========================================================================
    # Evaluate clones
    # =========================================================================

    evaluated_clones = []


    for clone in clone_population:

        clone_fitness = (
            evaluate_antibody(
                clone
            )
        )


        evaluated_clones.append(
            (
                clone,
                clone_fitness
            )
        )


    # =========================================================================
    # Combine parents + clones
    # =========================================================================

    combined = (

        evaluated_population
        +
        evaluated_clones
    )


    combined.sort(

        key=lambda item:
        item[1],

        reverse=True
    )


    # =========================================================================
    # Immune memory / survivor selection
    # =========================================================================

    next_population = []


    seen_keys = set()


    for antibody, fitness in combined:

        key = antibody_key(
            antibody
        )


        if key not in seen_keys:

            next_population.append(
                antibody.copy()
            )

            seen_keys.add(
                key
            )


        if len(
            next_population
        ) >= (
            AIS_POPULATION_SIZE
            -
            AIS_RANDOM_NEW
        ):

            break


    # =========================================================================
    # Diversity introduction
    # =========================================================================

    while len(
        next_population
    ) < AIS_POPULATION_SIZE:

        random_antibody = (
            create_random_antibody()
        )


        key = antibody_key(
            random_antibody
        )


        if key not in seen_keys:

            next_population.append(
                random_antibody
            )

            seen_keys.add(
                key
            )


    population = (
        next_population
    )


# =============================================================================
# 39. AIS OPTIMIZATION COMPLETE
# =============================================================================

print("\n" + "=" * 90)
print("AIS OPTIMIZATION COMPLETE")
print("=" * 90)


print(
    "\nBest AIS Fitness:"
)

print(
    global_best_fitness
)


print(
    "\nBest AIS Hyperparameters:"
)

for parameter, value in (
    global_best_antibody.items()
):

    print(
        f"{parameter}: {value}"
    )


# =============================================================================
# 40. SAVE OPTIMIZATION HISTORY
# =============================================================================

optimization_df = pd.DataFrame(
    optimization_history
)


optimization_df.to_csv(

    AIS_OPTIMIZATION_HISTORY_PATH,

    index=False
)


print(
    "\nAIS optimization history saved:"
)

print(
    AIS_OPTIMIZATION_HISTORY_PATH
)


# =============================================================================
# 41. OPTIMIZATION GRAPH
# =============================================================================

plt.figure(
    figsize=(10, 6)
)


plt.plot(

    optimization_df[
        "Generation"
    ],

    optimization_df[
        "Best_Fitness"
    ],

    marker="o",

    label="Generation Best"
)


plt.plot(

    optimization_df[
        "Generation"
    ],

    optimization_df[
        "Average_Fitness"
    ],

    marker="s",

    label="Generation Average"
)


plt.plot(

    optimization_df[
        "Generation"
    ],

    optimization_df[
        "Global_Best_Fitness"
    ],

    marker="^",

    linestyle="--",

    label="Global Best"
)


plt.xlabel(
    "AIS Generation"
)

plt.ylabel(
    "Cross-Validation Accuracy"
)

plt.title(
    "AIS Optimization Progress"
)

plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    AIS_OPTIMIZATION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# =============================================================================
# 42. TRAIN FINAL AIS-OPTIMIZED MODEL
# =============================================================================

print("\n" + "=" * 90)
print("TRAINING FINAL AIS-OPTIMIZED MODEL")
print("=" * 90)


ais_model = (
    create_pipeline_from_antibody(
        global_best_antibody
    )
)


ais_model.fit(
    X_train,
    y_train
)


# =============================================================================
# 43. AIS PREDICTIONS
# =============================================================================

ais_predictions = (
    ais_model.predict(
        X_test
    )
)


ais_probabilities = (
    ais_model.predict_proba(
        X_test
    )
)


# =============================================================================
# 44. AIS METRICS
# =============================================================================

ais_accuracy = accuracy_score(
    y_test,
    ais_predictions
)


ais_precision = precision_score(
    y_test,
    ais_predictions,
    average="weighted",
    zero_division=0
)


ais_recall = recall_score(
    y_test,
    ais_predictions,
    average="weighted",
    zero_division=0
)


ais_f1 = f1_score(
    y_test,
    ais_predictions,
    average="weighted",
    zero_division=0
)


print("\n" + "=" * 90)
print("AIS MODEL PERFORMANCE")
print("=" * 90)


print(
    "\nAccuracy:",
    f"{ais_accuracy * 100:.2f}%"
)

print(
    "Precision:",
    f"{ais_precision * 100:.2f}%"
)

print(
    "Recall:",
    f"{ais_recall * 100:.2f}%"
)

print(
    "F1 Score:",
    f"{ais_f1 * 100:.2f}%"
)


print(
    "\nClassification Report:"
)


print(

    classification_report(

        y_test,

        ais_predictions,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Low",
            "Medium",
            "High"
        ],

        zero_division=0
    )

)


# =============================================================================
# 45. SAVE AIS MODEL
# =============================================================================

with open(
    AIS_MODEL_PATH,
    "wb"
) as file:

    pickle.dump(
        ais_model,
        file
    )


print(
    "\nAIS model saved:"
)

print(
    AIS_MODEL_PATH
)


# =============================================================================
# 46. SAVE PROCESSED DATA
# =============================================================================

working_df.to_csv(

    AIS_PROCESSED_PATH,

    index=False
)


# =============================================================================
# 47. CREATE AIS PREDICTION CSV
# =============================================================================

prediction_df = (

    X_test

    .reset_index(
        drop=True
    )

    .copy()
)


actual_array = np.asarray(
    y_test
)


prediction_array = np.asarray(
    ais_predictions
)


prediction_df[
    "Actual_Class_Code"
] = actual_array


prediction_df[
    "Actual_Class"
] = [

    reverse_mapping[
        int(value)
    ]

    for value
    in actual_array
]


prediction_df[
    "AIS_Prediction_Code"
] = prediction_array


prediction_df[
    "AIS_Prediction"
] = [

    reverse_mapping[
        int(value)
    ]

    for value
    in prediction_array
]


prediction_df[
    "AIS_Correct"
] = (

    actual_array
    ==
    prediction_array
)


# =============================================================================
# 48. ADD PROBABILITIES
# =============================================================================

probability_classes = (
    ais_model
    .named_steps[
        "classifier"
    ]
    .classes_
)


for probability_index, class_id in enumerate(
    probability_classes
):

    class_name = (
        reverse_mapping[
            int(class_id)
        ]
    )


    prediction_df[
        f"AIS_Probability_{class_name}"
    ] = ais_probabilities[
        :,
        probability_index
    ]


# =============================================================================
# 49. SAVE PREDICTION CSV
# =============================================================================

prediction_df.to_csv(

    AIS_PREDICTION_PATH,

    index=False
)


print(
    "\nAIS prediction CSV:"
)

print(
    AIS_PREDICTION_PATH
)


print(
    "\nPrediction Preview:"
)

print(
    prediction_df.head(10)
)


# =============================================================================
# 50. RESULT CSV
# =============================================================================

result_df = pd.DataFrame(
    {

        "Method": [
            "AIS-Optimized Random Forest"
        ],

        "AIS_CV_Best_Accuracy": [
            global_best_fitness
        ],

        "Test_Accuracy": [
            ais_accuracy
        ],

        "Test_Accuracy_Percentage": [
            ais_accuracy * 100
        ],

        "Precision": [
            ais_precision
        ],

        "Precision_Percentage": [
            ais_precision * 100
        ],

        "Recall": [
            ais_recall
        ],

        "Recall_Percentage": [
            ais_recall * 100
        ],

        "F1_Score": [
            ais_f1
        ],

        "F1_Percentage": [
            ais_f1 * 100
        ],

        "n_estimators": [
            global_best_antibody[
                "n_estimators"
            ]
        ],

        "max_depth": [
            global_best_antibody[
                "max_depth"
            ]
        ],

        "min_samples_split": [
            global_best_antibody[
                "min_samples_split"
            ]
        ],

        "min_samples_leaf": [
            global_best_antibody[
                "min_samples_leaf"
            ]
        ],

        "max_features": [
            str(
                global_best_antibody[
                    "max_features"
                ]
            )
        ]
    }
)


result_df.to_csv(

    AIS_RESULT_PATH,

    index=False
)


print(
    "\nAIS result CSV:"
)

print(
    AIS_RESULT_PATH
)


print(
    "\nAIS Result:"
)

print(
    result_df
)


# =============================================================================
# 51. AIS ACCURACY GRAPH
# =============================================================================

plt.figure(
    figsize=(8, 6)
)


accuracy_values = [

    global_best_fitness * 100,

    ais_accuracy * 100
]


accuracy_labels = [

    "AIS CV Accuracy",

    "AIS Test Accuracy"
]


bars = plt.bar(

    accuracy_labels,

    accuracy_values
)


plt.ylabel(
    "Accuracy (%)"
)


plt.title(
    "AIS Model Accuracy"
)


plt.ylim(
    0,
    100
)


for bar, value in zip(
    bars,
    accuracy_values
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        value + 1,

        f"{value:.2f}%",

        ha="center",

        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    AIS_ACCURACY_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# =============================================================================
# 52. AIS PREDICTION GRAPH
# =============================================================================

number_of_samples = min(
    50,
    len(actual_array)
)


sample_numbers = np.arange(
    1,
    number_of_samples + 1
)


plt.figure(
    figsize=(14, 7)
)


plt.plot(

    sample_numbers,

    actual_array[
        :number_of_samples
    ],

    marker="o",

    linewidth=2,

    label="Actual"
)


plt.plot(

    sample_numbers,

    prediction_array[
        :number_of_samples
    ],

    marker="s",

    linestyle="--",

    linewidth=2,

    label="AIS Prediction"
)


plt.xlabel(
    "Test Sample"
)


plt.ylabel(
    "Production Performance"
)


plt.title(
    "Actual vs AIS Predicted Coal Mine Performance"
)


plt.yticks(

    [
        0,
        1,
        2
    ],

    [
        "Low",
        "Medium",
        "High"
    ]
)


plt.legend()

plt.grid(
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    AIS_PREDICTION_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# =============================================================================
# 53. AIS RESULT GRAPH
# =============================================================================

metrics = [

    "Accuracy",

    "Precision",

    "Recall",

    "F1 Score"
]


scores = [

    ais_accuracy * 100,

    ais_precision * 100,

    ais_recall * 100,

    ais_f1 * 100
]


plt.figure(
    figsize=(10, 6)
)


bars = plt.bar(
    metrics,
    scores
)


plt.xlabel(
    "Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "AIS-Optimized Model Performance"
)


plt.ylim(
    0,
    100
)


for bar, score in zip(
    bars,
    scores
):

    plt.text(

        bar.get_x()
        +
        bar.get_width() / 2,

        score + 1,

        f"{score:.2f}%",

        ha="center",

        fontweight="bold"
    )


plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    AIS_RESULT_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# =============================================================================
# 54. AIS COMPARISON GRAPH
# =============================================================================
#
# Compare initial/default Random Forest against AIS optimized Random Forest.
# =============================================================================

print("\n" + "=" * 90)
print("TRAINING BASELINE MODEL FOR AIS COMPARISON")
print("=" * 90)


baseline_antibody = {

    "n_estimators": 300,

    "max_depth": 12,

    "min_samples_split": 2,

    "min_samples_leaf": 1,

    "max_features": "sqrt"
}


baseline_model = (
    create_pipeline_from_antibody(
        baseline_antibody
    )
)


baseline_model.fit(
    X_train,
    y_train
)


baseline_predictions = (
    baseline_model.predict(
        X_test
    )
)


baseline_accuracy = accuracy_score(
    y_test,
    baseline_predictions
)


baseline_precision = precision_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)


baseline_recall = recall_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)


baseline_f1 = f1_score(
    y_test,
    baseline_predictions,
    average="weighted",
    zero_division=0
)


baseline_scores = [

    baseline_accuracy * 100,

    baseline_precision * 100,

    baseline_recall * 100,

    baseline_f1 * 100
]


ais_scores = [

    ais_accuracy * 100,

    ais_precision * 100,

    ais_recall * 100,

    ais_f1 * 100
]


x = np.arange(
    len(metrics)
)


width = 0.35


plt.figure(
    figsize=(11, 7)
)


baseline_bars = plt.bar(

    x - width / 2,

    baseline_scores,

    width,

    label="Baseline Random Forest"
)


ais_bars = plt.bar(

    x + width / 2,

    ais_scores,

    width,

    label="AIS-Optimized Random Forest"
)


plt.xlabel(
    "Evaluation Metric"
)


plt.ylabel(
    "Score (%)"
)


plt.title(
    "Baseline vs AIS-Optimized Model"
)


plt.xticks(
    x,
    metrics
)


plt.ylim(
    0,
    100
)


plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)


for bars in [

    baseline_bars,

    ais_bars

]:

    for bar in bars:

        height = (
            bar.get_height()
        )


        plt.text(

            bar.get_x()
            +
            bar.get_width() / 2,

            height + 0.5,

            f"{height:.1f}",

            ha="center",

            fontsize=8
        )


plt.tight_layout()


plt.savefig(

    AIS_COMPARISON_GRAPH_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# =============================================================================
# 55. AIS HEATMAP
# =============================================================================

ais_cm = confusion_matrix(

    y_test,

    ais_predictions,

    labels=[
        0,
        1,
        2
    ]
)


print(
    "\nAIS Confusion Matrix:"
)

print(
    ais_cm
)


class_names = [

    "Low",

    "Medium",

    "High"
]


plt.figure(
    figsize=(8, 7)
)


image = plt.imshow(

    ais_cm,

    interpolation="nearest",

    cmap="Blues"
)


plt.colorbar(
    image
)


plt.title(
    "AIS-Optimized Model Confusion Matrix"
)


plt.xlabel(
    "Predicted Class"
)


plt.ylabel(
    "Actual Class"
)


tick_positions = np.arange(
    len(class_names)
)


plt.xticks(
    tick_positions,
    class_names
)


plt.yticks(
    tick_positions,
    class_names
)


threshold = (

    ais_cm.max() / 2

    if ais_cm.size > 0

    else 0
)


for row in range(
    ais_cm.shape[0]
):

    for column in range(
        ais_cm.shape[1]
    ):

        value = ais_cm[
            row,
            column
        ]


        color = (

            "white"

            if value > threshold

            else "black"
        )


        plt.text(

            column,

            row,

            str(value),

            ha="center",

            va="center",

            color=color,

            fontsize=14,

            fontweight="bold"
        )


plt.tight_layout()


plt.savefig(

    AIS_HEATMAP_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# =============================================================================
# 56. CLASS DISTRIBUTION GRAPH
# =============================================================================

actual_counts = [

    np.sum(
        actual_array == class_id
    )

    for class_id in [
        0,
        1,
        2
    ]
]


predicted_counts = [

    np.sum(
        prediction_array == class_id
    )

    for class_id in [
        0,
        1,
        2
    ]
]


x = np.arange(
    len(class_names)
)


width = 0.35


plt.figure(
    figsize=(10, 6)
)


plt.bar(

    x - width / 2,

    actual_counts,

    width,

    label="Actual"
)


plt.bar(

    x + width / 2,

    predicted_counts,

    width,

    label="AIS Predicted"
)


plt.xlabel(
    "Performance Class"
)


plt.ylabel(
    "Number of Samples"
)


plt.title(
    "Actual vs AIS Predicted Class Distribution"
)


plt.xticks(
    x,
    class_names
)


plt.legend()

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()


plt.savefig(

    AIS_CLASS_DISTRIBUTION_PATH,

    dpi=300,

    bbox_inches="tight"
)


plt.show()


# =============================================================================
# 57. SAVE YAML CONFIG
# =============================================================================

yaml_config = {

    "project": {

        "name":
            "AI-Based Coal Mine Production Analysis "
            "and Performance Classification",

        "optimization":
            "Artificial Immune System",

        "ais_algorithm":
            "Clonal Selection Algorithm"
    },


    "dataset": {

        "path":
            CSV_PATH,

        "mine_column":
            mine_column,

        "company_column":
            company_column,

        "production_column":
            production_column,

        "features":
            feature_columns
    },


    "target": {

        "classes": {

            0: "Low",

            1: "Medium",

            2: "High"
        },

        "low_medium_threshold":
            float(
                q33
            ),

        "medium_high_threshold":
            float(
                q66
            )
    },


    "ais": {

        "population_size":
            AIS_POPULATION_SIZE,

        "generations":
            AIS_GENERATIONS,

        "clone_factor":
            AIS_CLONE_FACTOR,

        "elite_size":
            AIS_ELITE_SIZE,

        "random_new_antibodies":
            AIS_RANDOM_NEW,

        "mutation_rate":
            AIS_MUTATION_RATE,

        "best_cv_fitness":
            float(
                global_best_fitness
            ),

        "best_parameters":
            {

                key:
                    (
                        str(value)
                        if value is None
                        else value
                    )

                for key, value
                in global_best_antibody.items()
            }
    },


    "performance": {

        "accuracy":
            float(
                ais_accuracy
            ),

        "precision":
            float(
                ais_precision
            ),

        "recall":
            float(
                ais_recall
            ),

        "f1_score":
            float(
                ais_f1
            )
    }
}


with open(

    AIS_CONFIG_PATH,

    "w",

    encoding="utf-8"

) as file:

    yaml.safe_dump(

        yaml_config,

        file,

        sort_keys=False,

        allow_unicode=True
    )


# =============================================================================
# 58. JSON SERIALIZATION HELPER
# =============================================================================

def serializable(
    value
):

    if isinstance(
        value,
        np.integer
    ):

        return int(
            value
        )


    if isinstance(
        value,
        np.floating
    ):

        return float(
            value
        )


    if isinstance(
        value,
        np.ndarray
    ):

        return value.tolist()


    if isinstance(
        value,
        dict
    ):

        return {

            str(key):
                serializable(
                    item
                )

            for key, item
            in value.items()
        }


    if isinstance(
        value,
        list
    ):

        return [

            serializable(
                item
            )

            for item in value
        ]


    return value


# =============================================================================
# 59. CLASSIFICATION REPORT DICTIONARY
# =============================================================================

ais_report = classification_report(

    y_test,

    ais_predictions,

    labels=[
        0,
        1,
        2
    ],

    target_names=[
        "Low",
        "Medium",
        "High"
    ],

    zero_division=0,

    output_dict=True
)


# =============================================================================
# 60. SAVE JSON METADATA
# =============================================================================

metadata = {

    "project":
        "AI-Based Coal Mine Production Analysis "
        "and Performance Classification",

    "method":
        "Artificial Immune System optimized Random Forest",

    "ais_algorithm":
        "Clonal Selection Algorithm",

    "records":
        int(
            len(
                working_df
            )
        ),

    "features":
        feature_columns,

    "best_antibody":
        global_best_antibody,

    "best_cross_validation_accuracy":
        global_best_fitness,

    "test_metrics": {

        "accuracy":
            ais_accuracy,

        "precision":
            ais_precision,

        "recall":
            ais_recall,

        "f1_score":
            ais_f1
    },

    "baseline_metrics": {

        "accuracy":
            baseline_accuracy,

        "precision":
            baseline_precision,

        "recall":
            baseline_recall,

        "f1_score":
            baseline_f1
    },

    "confusion_matrix":
        ais_cm,

    "classification_report":
        ais_report
}


with open(

    AIS_METADATA_PATH,

    "w",

    encoding="utf-8"

) as file:

    json.dump(

        serializable(
            metadata
        ),

        file,

        indent=4,

        ensure_ascii=False
    )


# =============================================================================
# 61. VERIFY FILES
# =============================================================================

print("\n" + "=" * 90)
print("VERIFYING AIS OUTPUT FILES")
print("=" * 90)


generated_files = [

    AIS_MODEL_PATH,

    AIS_CONFIG_PATH,

    AIS_METADATA_PATH,

    AIS_PROCESSED_PATH,

    AIS_PREDICTION_PATH,

    AIS_RESULT_PATH,

    AIS_OPTIMIZATION_HISTORY_PATH,

    AIS_ACCURACY_GRAPH_PATH,

    AIS_PREDICTION_GRAPH_PATH,

    AIS_RESULT_GRAPH_PATH,

    AIS_COMPARISON_GRAPH_PATH,

    AIS_HEATMAP_PATH,

    AIS_OPTIMIZATION_GRAPH_PATH,

    AIS_CLASS_DISTRIBUTION_PATH
]


for path in generated_files:

    if os.path.exists(
        path
    ):

        size = os.path.getsize(
            path
        )


        print(

            f"[OK] "
            f"{os.path.basename(path)} "
            f"({size:,} bytes)"
        )

    else:

        print(

            f"[MISSING] "
            f"{os.path.basename(path)}"
        )


# =============================================================================
# 62. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 90)
print("AIS PROJECT COMPLETED SUCCESSFULLY")
print("=" * 90)


print(
    f"""

PROJECT
=======

AI-Based Coal Mine Production Analysis
and Performance Classification


AIS METHOD
==========

Artificial Immune System
Clonal Selection Algorithm


BEST ANTIBODY
=============

n_estimators:
{global_best_antibody['n_estimators']}

max_depth:
{global_best_antibody['max_depth']}

min_samples_split:
{global_best_antibody['min_samples_split']}

min_samples_leaf:
{global_best_antibody['min_samples_leaf']}

max_features:
{global_best_antibody['max_features']}


BEST CROSS-VALIDATION FITNESS
=============================

{global_best_fitness * 100:.2f}%


BASELINE RANDOM FOREST
======================

Accuracy:
{baseline_accuracy * 100:.2f}%

Precision:
{baseline_precision * 100:.2f}%

Recall:
{baseline_recall * 100:.2f}%

F1 Score:
{baseline_f1 * 100:.2f}%


AIS-OPTIMIZED RANDOM FOREST
===========================

Accuracy:
{ais_accuracy * 100:.2f}%

Precision:
{ais_precision * 100:.2f}%

Recall:
{ais_recall * 100:.2f}%

F1 Score:
{ais_f1 * 100:.2f}%


GENERATED FILES
===============

1. ais_model.pkl

2. ais_config.yaml

3. ais_metadata.json

4. ais_processed_data.csv

5. ais_prediction.csv

6. ais_result.csv

7. ais_optimization_history.csv

8. ais_accuracy_graph.png

9. ais_prediction_graph.png

10. ais_result_graph.png

11. ais_comparison_graph.png

12. ais_heatmap.png

13. ais_optimization_graph.png

14. ais_class_distribution_graph.png


OUTPUT DIRECTORY
================

{PROJECT_DIR}

"""
)


print("=" * 90)
print("ALL AIS RESULTS SAVED WITH ais_ PREFIX")
print("=" * 90)
```
