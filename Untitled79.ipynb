```python
# =============================================================================
# AI-BASED COAL MINE PRODUCTION ANALYSIS AND PERFORMANCE CLASSIFICATION
# FIXED COMPLETE VERSION
# =============================================================================

# =============================================================================
# 1. IMPORT LIBRARIES
# =============================================================================

import os
import re
import json
import pickle
import warnings

import numpy as np
import pandas as pd
import yaml

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import (
    Input,
    Dense,
    Dropout
)

from tensorflow.keras.callbacks import EarlyStopping


warnings.filterwarnings("ignore")

np.random.seed(42)
tf.random.set_seed(42)


# =============================================================================
# 2. PROJECT PATH
# =============================================================================

PROJECT_DIR = (
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Coal Mine Production Analysis and Performance Classification"
)

CSV_PATH = os.path.join(
    PROJECT_DIR,
    "rs_session-241_au482_1.1.csv"
)


# =============================================================================
# 3. OUTPUT PATHS
# =============================================================================

PKL_MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "coal_production_model.pkl"
)

H5_MODEL_PATH = os.path.join(
    PROJECT_DIR,
    "coal_production_model.h5"
)

YAML_PATH = os.path.join(
    PROJECT_DIR,
    "coal_production_config.yaml"
)

JSON_PATH = os.path.join(
    PROJECT_DIR,
    "coal_production_metadata.json"
)

PREPROCESSOR_PATH = os.path.join(
    PROJECT_DIR,
    "coal_preprocessor.pkl"
)

PROCESSED_CSV_PATH = os.path.join(
    PROJECT_DIR,
    "processed_coal_production_data.csv"
)

TEST_PREDICTION_PATH = os.path.join(
    PROJECT_DIR,
    "test_predictions.csv"
)


os.makedirs(
    PROJECT_DIR,
    exist_ok=True
)


# =============================================================================
# 4. DISPLAY HEADER
# =============================================================================

print("=" * 80)

print(
    "AI-BASED COAL MINE PRODUCTION ANALYSIS "
    "AND PERFORMANCE CLASSIFICATION"
)

print("=" * 80)


# =============================================================================
# 5. CHECK DATASET
# =============================================================================

if not os.path.exists(CSV_PATH):

    raise FileNotFoundError(
        f"\nDataset not found:\n{CSV_PATH}\n\n"
        "Please verify the dataset path."
    )


print("\nDataset found successfully.")

print("\nDataset path:")
print(CSV_PATH)


# =============================================================================
# 6. LOAD CSV USING DIFFERENT ENCODINGS
# =============================================================================

print("\n" + "=" * 80)
print("LOADING DATASET")
print("=" * 80)


encodings = [
    "utf-8",
    "utf-8-sig",
    "cp1252",
    "latin1"
]


df = None
used_encoding = None


for encoding in encodings:

    try:

        df = pd.read_csv(
            CSV_PATH,
            encoding=encoding
        )

        used_encoding = encoding

        break

    except UnicodeDecodeError:

        continue


if df is None:

    raise ValueError(
        "Unable to load the CSV file."
    )


print("\nDataset loaded successfully.")

print("\nEncoding:")
print(used_encoding)

print("\nOriginal shape:")
print(df.shape)


# =============================================================================
# 7. CLEAN COLUMN NAMES
# =============================================================================

print("\n" + "=" * 80)
print("CLEANING COLUMN NAMES")
print("=" * 80)


def clean_column_name(column):

    column = str(column)

    column = column.replace(
        "\n",
        " "
    )

    column = column.replace(
        "\r",
        " "
    )

    column = re.sub(
        r"\s+",
        " ",
        column
    )

    return column.strip()


df.columns = [
    clean_column_name(col)
    for col in df.columns
]


print("\nColumns after cleaning:")

for index, col in enumerate(
    df.columns,
    start=1
):

    print(
        f"{index}. {col}"
    )


# =============================================================================
# 8. FIX DUPLICATE COLUMN NAMES
# =============================================================================

print("\n" + "=" * 80)
print("CHECKING DUPLICATE COLUMN NAMES")
print("=" * 80)


def make_unique_columns(columns):

    seen = {}

    new_columns = []

    for column in columns:

        if column not in seen:

            seen[column] = 0

            new_columns.append(
                column
            )

        else:

            seen[column] += 1

            new_name = (
                f"{column}_{seen[column]}"
            )

            while new_name in seen:

                seen[column] += 1

                new_name = (
                    f"{column}_{seen[column]}"
                )

            seen[new_name] = 0

            new_columns.append(
                new_name
            )

    return new_columns


original_columns = list(
    df.columns
)


df.columns = make_unique_columns(
    df.columns
)


if original_columns != list(df.columns):

    print(
        "\nDuplicate column names were detected "
        "and automatically renamed."
    )

else:

    print(
        "\nNo duplicate column names detected."
    )


print("\nFinal unique columns:")

for index, col in enumerate(
    df.columns,
    start=1
):

    print(
        f"{index}. {col}"
    )


# =============================================================================
# 9. REMOVE EMPTY ROWS AND COLUMNS
# =============================================================================

print("\n" + "=" * 80)
print("REMOVING EMPTY DATA")
print("=" * 80)


before_shape = df.shape


df = df.dropna(
    axis=0,
    how="all"
)


df = df.dropna(
    axis=1,
    how="all"
)


print("\nBefore:")
print(before_shape)

print("\nAfter:")
print(df.shape)


# =============================================================================
# 10. DISPLAY SAMPLE
# =============================================================================

print("\n" + "=" * 80)
print("DATASET SAMPLE")
print("=" * 80)

print(
    df.head(10)
)


# =============================================================================
# 11. DATA TYPES
# =============================================================================

print("\n" + "=" * 80)
print("DATA TYPES")
print("=" * 80)

print(
    df.dtypes
)


# =============================================================================
# 12. FIND COLUMNS
# =============================================================================

def find_column(
    columns,
    keywords,
    exclude=None
):

    if exclude is None:

        exclude = []

    exclude = set(exclude)

    # ---------------------------------------------------------
    # First pass:
    # Strong keyword match
    # ---------------------------------------------------------

    for keyword in keywords:

        keyword_lower = (
            keyword.lower().strip()
        )

        for column in columns:

            if column in exclude:
                continue

            column_lower = (
                str(column)
                .lower()
                .strip()
            )

            if keyword_lower == column_lower:

                return column

    # ---------------------------------------------------------
    # Second pass:
    # Keyword contained inside column
    # ---------------------------------------------------------

    for keyword in keywords:

        keyword_lower = (
            keyword.lower().strip()
        )

        for column in columns:

            if column in exclude:
                continue

            column_lower = (
                str(column)
                .lower()
                .strip()
            )

            if keyword_lower in column_lower:

                return column

    return None


# =============================================================================
# 13. DETECT PRODUCTION COLUMN FIRST
# =============================================================================

print("\n" + "=" * 80)
print("DETECTING IMPORTANT COLUMNS")
print("=" * 80)


production_column = find_column(
    df.columns,
    [
        "quantity of coal production",
        "coal production quantity",
        "coal production",
        "production quantity",
        "production",
        "quantity"
    ]
)


# =============================================================================
# 14. FALLBACK NUMERIC DETECTION
# =============================================================================

if production_column is None:

    print(
        "\nProduction column was not identified "
        "from its name."
    )

    print(
        "Searching for the most numeric column..."
    )


    best_column = None

    best_numeric_ratio = -1


    for column in df.columns:

        values = (
            df[column]
            .astype(str)
            .str.replace(
                ",",
                "",
                regex=False
            )
            .str.replace(
                "₹",
                "",
                regex=False
            )
            .str.strip()
        )


        numeric_values = pd.to_numeric(
            values,
            errors="coerce"
        )


        numeric_ratio = (
            numeric_values
            .notna()
            .mean()
        )


        if numeric_ratio > best_numeric_ratio:

            best_numeric_ratio = (
                numeric_ratio
            )

            best_column = column


    production_column = (
        best_column
    )


print("\nDetected production column:")

print(
    production_column
)


# =============================================================================
# 15. DETECT MINE COLUMN
# =============================================================================

mine_column = find_column(
    df.columns,
    [
        "name of mine",
        "mine name",
        "name of mines",
        "mine"
    ],
    exclude=[
        production_column
    ]
)


# =============================================================================
# 16. DETECT COMPANY COLUMN
# =============================================================================

company_column = find_column(
    df.columns,
    [
        "coal company",
        "company name",
        "name of company",
        "company",
        "subsidiary",
        "region"
    ],
    exclude=[
        production_column,
        mine_column
    ]
)


# =============================================================================
# 17. POSITIONAL FALLBACK
# =============================================================================

available_columns = [
    col
    for col in df.columns
    if col != production_column
]


if mine_column is None:

    if len(available_columns) >= 1:

        mine_column = (
            available_columns[0]
        )


if company_column is None:

    remaining_columns = [
        col
        for col in available_columns
        if col != mine_column
    ]

    if len(remaining_columns) >= 1:

        company_column = (
            remaining_columns[0]
        )


print("\nDetected mine column:")
print(
    mine_column
)

print("\nDetected company/region column:")
print(
    company_column
)

print("\nDetected production column:")
print(
    production_column
)


# =============================================================================
# 18. VERIFY DIFFERENT COLUMNS
# =============================================================================

if production_column is None:

    raise ValueError(
        "Production column could not be identified."
    )


if mine_column == production_column:

    mine_column = None


if company_column == production_column:

    company_column = None


if (
    mine_column is not None
    and company_column == mine_column
):

    company_column = None


# =============================================================================
# 19. COPY DATA
# =============================================================================

working_df = df.copy()


# =============================================================================
# 20. SAFE SERIES CLEANING FUNCTION
# =============================================================================

def clean_categorical_series(series):

    """
    Clean categorical values safely.

    The input must be a Pandas Series.
    """

    series = series.copy()

    series = series.fillna(
        "Unknown"
    )

    series = series.astype(
        str
    )

    series = series.map(
        lambda x: x.strip()
    )

    invalid_values = [
        "",
        "nan",
        "NaN",
        "none",
        "None",
        "NULL",
        "null"
    ]

    series = series.replace(
        invalid_values,
        "Unknown"
    )

    return series


# =============================================================================
# 21. CLEAN MINE COLUMN
# =============================================================================

if mine_column is not None:

    working_df[
        mine_column
    ] = clean_categorical_series(
        working_df[
            mine_column
        ]
    )


# =============================================================================
# 22. CLEAN COMPANY COLUMN
# =============================================================================

if company_column is not None:

    working_df[
        company_column
    ] = clean_categorical_series(
        working_df[
            company_column
        ]
    )


# =============================================================================
# 23. CLEAN PRODUCTION COLUMN
# =============================================================================

print("\n" + "=" * 80)
print("CLEANING PRODUCTION VALUES")
print("=" * 80)


production_values = (
    working_df[
        production_column
    ]
    .astype(str)
    .map(
        lambda x: x.strip()
    )
)


production_values = (
    production_values
    .str.replace(
        ",",
        "",
        regex=False
    )
    .str.replace(
        "₹",
        "",
        regex=False
    )
)


# Keep digits, decimal point and minus sign

production_values = (
    production_values
    .str.replace(
        r"[^0-9.\-]",
        "",
        regex=True
    )
)


working_df[
    production_column
] = pd.to_numeric(
    production_values,
    errors="coerce"
)


# =============================================================================
# 24. REMOVE MISSING PRODUCTION
# =============================================================================

rows_before = len(
    working_df
)


working_df = (
    working_df
    .dropna(
        subset=[
            production_column
        ]
    )
    .copy()
)


rows_after = len(
    working_df
)


print(
    "\nRows removed because production "
    "was missing/invalid:"
)

print(
    rows_before - rows_after
)


# =============================================================================
# 25. REMOVE NEGATIVE PRODUCTION
# =============================================================================

negative_count = int(
    (
        working_df[
            production_column
        ] < 0
    ).sum()
)


print(
    "\nNegative production values:"
)

print(
    negative_count
)


working_df = (
    working_df[
        working_df[
            production_column
        ] >= 0
    ]
    .copy()
)


# =============================================================================
# 26. RESET INDEX
# =============================================================================

working_df.reset_index(
    drop=True,
    inplace=True
)


if len(working_df) < 10:

    raise ValueError(
        "Too few valid records remain "
        "after data cleaning."
    )


# =============================================================================
# 27. PRODUCTION STATISTICS
# =============================================================================

print("\n" + "=" * 80)
print("PRODUCTION STATISTICS")
print("=" * 80)


print(
    working_df[
        production_column
    ].describe()
)


# =============================================================================
# 28. CREATE LOW / MEDIUM / HIGH TARGET
# =============================================================================

print("\n" + "=" * 80)
print("CREATING PRODUCTION PERFORMANCE CLASSES")
print("=" * 80)


TARGET_COLUMN = (
    "Production_Performance"
)


# Store actual quantile thresholds

q33 = float(
    working_df[
        production_column
    ].quantile(
        1 / 3
    )
)


q66 = float(
    working_df[
        production_column
    ].quantile(
        2 / 3
    )
)


print(
    "\n33.33% quantile:"
)

print(
    q33
)


print(
    "\n66.67% quantile:"
)

print(
    q66
)


# =============================================================================
# 29. CREATE TARGET USING QCUT
# =============================================================================

try:

    working_df[
        TARGET_COLUMN
    ] = pd.qcut(

        working_df[
            production_column
        ],

        q=3,

        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


except ValueError:

    print(
        "\nDuplicate quantile boundaries detected."
    )

    print(
        "Using ranked production values."
    )


    ranked_production = (
        working_df[
            production_column
        ]
        .rank(
            method="first"
        )
    )


    working_df[
        TARGET_COLUMN
    ] = pd.qcut(

        ranked_production,

        q=3,

        labels=[
            "Low",
            "Medium",
            "High"
        ]
    )


# =============================================================================
# 30. TARGET MAPPING
# =============================================================================

class_mapping = {
    "Low": 0,
    "Medium": 1,
    "High": 2
}


reverse_mapping = {
    0: "Low",
    1: "Medium",
    2: "High"
}


working_df[
    "Target"
] = (
    working_df[
        TARGET_COLUMN
    ]
    .astype(str)
    .map(
        class_mapping
    )
)


print(
    "\nProduction class distribution:"
)

print(
    working_df[
        TARGET_COLUMN
    ].value_counts()
)


# =============================================================================
# 31. BUILD FEATURE LIST
# =============================================================================

print("\n" + "=" * 80)
print("BUILDING FEATURE LIST")
print("=" * 80)


feature_columns = []


if mine_column is not None:

    if mine_column not in feature_columns:

        feature_columns.append(
            mine_column
        )


if company_column is not None:

    if company_column not in feature_columns:

        feature_columns.append(
            company_column
        )


# Never use production as input because it created the target

feature_columns = [
    col
    for col in feature_columns
    if col != production_column
]


# Remove target columns if accidentally included

feature_columns = [
    col
    for col in feature_columns
    if col not in [
        TARGET_COLUMN,
        "Target"
    ]
]


# Remove duplicate names while preserving order

feature_columns = list(
    dict.fromkeys(
        feature_columns
    )
)


print(
    "\nFinal feature columns:"
)

for col in feature_columns:

    print(
        "-",
        col
    )


if len(feature_columns) == 0:

    raise ValueError(
        "No valid feature columns were found."
    )


# =============================================================================
# 32. CREATE X AND y
# =============================================================================

X = (
    working_df[
        feature_columns
    ]
    .copy()
)


y = (
    working_df[
        "Target"
    ]
    .copy()
)


# =============================================================================
# 33. EXTRA DUPLICATE-COLUMN SAFETY CHECK
# =============================================================================

if X.columns.duplicated().any():

    print(
        "\nDuplicate feature columns detected."
    )

    print(
        "Removing duplicates..."
    )


    X = X.loc[
        :,
        ~X.columns.duplicated()
    ].copy()


feature_columns = (
    X.columns.tolist()
)


# =============================================================================
# 34. SAFE FEATURE CLEANING
# =============================================================================

print("\n" + "=" * 80)
print("CLEANING INPUT FEATURES")
print("=" * 80)


for column_index in range(
    X.shape[1]
):

    # iloc guarantees that we obtain
    # exactly one Series even if names are unusual.

    series = X.iloc[
        :,
        column_index
    ]


    cleaned_series = (
        clean_categorical_series(
            series
        )
    )


    X.iloc[
        :,
        column_index
    ] = cleaned_series


print(
    "\nFeature cleaning completed."
)


print(
    "\nX shape:"
)

print(
    X.shape
)


print(
    "\ny shape:"
)

print(
    y.shape
)


# =============================================================================
# 35. CLASS COUNTS
# =============================================================================

class_counts = (
    y.value_counts()
)


print(
    "\nClass counts:"
)

print(
    class_counts
)


if class_counts.min() < 2:

    raise ValueError(
        "At least one target class contains fewer "
        "than two samples. Stratified splitting "
        "cannot be performed."
    )


# =============================================================================
# 36. TRAIN TEST SPLIT
# =============================================================================

print("\n" + "=" * 80)
print("TRAIN / TEST SPLIT")
print("=" * 80)


X_train, X_test, y_train, y_test = (
    train_test_split(

        X,
        y,

        test_size=0.20,

        random_state=42,

        stratify=y
    )
)


print(
    "\nTraining records:"
)

print(
    len(X_train)
)


print(
    "\nTesting records:"
)

print(
    len(X_test)
)


# =============================================================================
# 37. CREATE ONE-HOT ENCODER
# =============================================================================

def create_one_hot_encoder():

    try:

        return OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )

    except TypeError:

        return OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )


# =============================================================================
# 38. RANDOM FOREST PREPROCESSOR
# =============================================================================

rf_encoder = (
    create_one_hot_encoder()
)


rf_preprocessor = ColumnTransformer(

    transformers=[

        (
            "categorical",

            rf_encoder,

            feature_columns
        )

    ],

    remainder="drop"
)


# =============================================================================
# 39. RANDOM FOREST
# =============================================================================

rf_classifier = RandomForestClassifier(

    n_estimators=300,

    max_depth=12,

    min_samples_split=2,

    min_samples_leaf=1,

    class_weight="balanced",

    random_state=42,

    n_jobs=-1
)


# =============================================================================
# 40. RANDOM FOREST PIPELINE
# =============================================================================

rf_pipeline = Pipeline(

    steps=[

        (
            "preprocessor",
            rf_preprocessor
        ),

        (
            "classifier",
            rf_classifier
        )

    ]
)


# =============================================================================
# 41. TRAIN RANDOM FOREST
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING RANDOM FOREST")
print("=" * 80)


rf_pipeline.fit(
    X_train,
    y_train
)


print(
    "\nRandom Forest training completed."
)


# =============================================================================
# 42. RANDOM FOREST PREDICTIONS
# =============================================================================

rf_predictions = (
    rf_pipeline.predict(
        X_test
    )
)


rf_accuracy = (
    accuracy_score(
        y_test,
        rf_predictions
    )
)


print(
    "\nRandom Forest Accuracy:"
)

print(
    f"{rf_accuracy:.4f}"
)

print(
    f"{rf_accuracy * 100:.2f}%"
)


# =============================================================================
# 43. RANDOM FOREST REPORT
# =============================================================================

rf_report_text = (
    classification_report(

        y_test,

        rf_predictions,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Low",
            "Medium",
            "High"
        ],

        zero_division=0
    )
)


print(
    "\nRandom Forest Classification Report:"
)

print(
    rf_report_text
)


rf_report = (
    classification_report(

        y_test,

        rf_predictions,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Low",
            "Medium",
            "High"
        ],

        zero_division=0,

        output_dict=True
    )
)


# =============================================================================
# 44. RANDOM FOREST CONFUSION MATRIX
# =============================================================================

rf_cm = confusion_matrix(

    y_test,

    rf_predictions,

    labels=[
        0,
        1,
        2
    ]
)


print(
    "\nRandom Forest Confusion Matrix:"
)

print(
    rf_cm
)


# =============================================================================
# 45. SAVE RANDOM FOREST PKL
# =============================================================================

print("\n" + "=" * 80)
print("SAVING RANDOM FOREST MODEL")
print("=" * 80)


with open(
    PKL_MODEL_PATH,
    "wb"
) as file:

    pickle.dump(
        rf_pipeline,
        file
    )


print(
    "\nPKL saved:"
)

print(
    PKL_MODEL_PATH
)


# =============================================================================
# 46. CREATE SEPARATE NN PREPROCESSOR
# =============================================================================

print("\n" + "=" * 80)
print("PREPARING NEURAL NETWORK DATA")
print("=" * 80)


nn_encoder = (
    create_one_hot_encoder()
)


nn_preprocessor = ColumnTransformer(

    transformers=[

        (
            "categorical",

            nn_encoder,

            feature_columns
        )

    ],

    remainder="drop"
)


# =============================================================================
# 47. TRANSFORM DATA
# =============================================================================

X_train_processed = (
    nn_preprocessor.fit_transform(
        X_train
    )
)


X_test_processed = (
    nn_preprocessor.transform(
        X_test
    )
)


X_train_processed = np.asarray(
    X_train_processed,
    dtype=np.float32
)


X_test_processed = np.asarray(
    X_test_processed,
    dtype=np.float32
)


print(
    "\nProcessed training shape:"
)

print(
    X_train_processed.shape
)


print(
    "\nProcessed testing shape:"
)

print(
    X_test_processed.shape
)


# =============================================================================
# 48. SCALE NN DATA
# =============================================================================

scaler = StandardScaler()


X_train_scaled = (
    scaler.fit_transform(
        X_train_processed
    )
)


X_test_scaled = (
    scaler.transform(
        X_test_processed
    )
)


X_train_scaled = np.asarray(
    X_train_scaled,
    dtype=np.float32
)


X_test_scaled = np.asarray(
    X_test_scaled,
    dtype=np.float32
)


# =============================================================================
# 49. BUILD NEURAL NETWORK
# =============================================================================

print("\n" + "=" * 80)
print("BUILDING NEURAL NETWORK")
print("=" * 80)


input_dimension = (
    X_train_scaled.shape[1]
)


nn_model = Sequential(
    [

        Input(
            shape=(
                input_dimension,
            )
        ),

        Dense(
            128,
            activation="relu"
        ),

        Dropout(
            0.30
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dropout(
            0.20
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            3,
            activation="softmax"
        )

    ]
)


nn_model.compile(

    optimizer="adam",

    loss=(
        "sparse_categorical_crossentropy"
    ),

    metrics=[
        "accuracy"
    ]
)


nn_model.summary()


# =============================================================================
# 50. EARLY STOPPING
# =============================================================================

early_stopping = EarlyStopping(

    monitor="val_loss",

    patience=15,

    restore_best_weights=True,

    verbose=1
)


# =============================================================================
# 51. TRAIN NEURAL NETWORK
# =============================================================================

print("\n" + "=" * 80)
print("TRAINING NEURAL NETWORK")
print("=" * 80)


history = nn_model.fit(

    X_train_scaled,

    y_train.to_numpy(
        dtype=np.int32
    ),

    validation_split=0.20,

    epochs=150,

    batch_size=16,

    callbacks=[
        early_stopping
    ],

    verbose=1
)


# =============================================================================
# 52. EVALUATE NEURAL NETWORK
# =============================================================================

nn_loss, nn_accuracy = (
    nn_model.evaluate(

        X_test_scaled,

        y_test.to_numpy(
            dtype=np.int32
        ),

        verbose=0
    )
)


print(
    "\nNeural Network Loss:"
)

print(
    f"{nn_loss:.4f}"
)


print(
    "\nNeural Network Accuracy:"
)

print(
    f"{nn_accuracy:.4f}"
)

print(
    f"{nn_accuracy * 100:.2f}%"
)


# =============================================================================
# 53. NN PREDICTIONS
# =============================================================================

nn_probabilities = (
    nn_model.predict(
        X_test_scaled,
        verbose=0
    )
)


nn_predictions = np.argmax(
    nn_probabilities,
    axis=1
)


# =============================================================================
# 54. NN CLASSIFICATION REPORT
# =============================================================================

nn_report_text = (
    classification_report(

        y_test,

        nn_predictions,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Low",
            "Medium",
            "High"
        ],

        zero_division=0
    )
)


print(
    "\nNeural Network Classification Report:"
)

print(
    nn_report_text
)


nn_report = (
    classification_report(

        y_test,

        nn_predictions,

        labels=[
            0,
            1,
            2
        ],

        target_names=[
            "Low",
            "Medium",
            "High"
        ],

        zero_division=0,

        output_dict=True
    )
)


# =============================================================================
# 55. NN CONFUSION MATRIX
# =============================================================================

nn_cm = confusion_matrix(

    y_test,

    nn_predictions,

    labels=[
        0,
        1,
        2
    ]
)


print(
    "\nNeural Network Confusion Matrix:"
)

print(
    nn_cm
)


# =============================================================================
# 56. SAVE H5
# =============================================================================

print("\n" + "=" * 80)
print("SAVING H5 MODEL")
print("=" * 80)


nn_model.save(
    H5_MODEL_PATH
)


print(
    "\nH5 model saved:"
)

print(
    H5_MODEL_PATH
)


# =============================================================================
# 57. SAVE NN PREPROCESSING OBJECTS
# =============================================================================

preprocessing_objects = {

    "preprocessor":
        nn_preprocessor,

    "scaler":
        scaler,

    "feature_columns":
        feature_columns,

    "mine_column":
        mine_column,

    "company_column":
        company_column,

    "production_column":
        production_column,

    "class_mapping":
        class_mapping,

    "reverse_mapping":
        reverse_mapping,

    "low_medium_threshold":
        q33,

    "medium_high_threshold":
        q66
}


with open(
    PREPROCESSOR_PATH,
    "wb"
) as file:

    pickle.dump(
        preprocessing_objects,
        file
    )


print(
    "\nPreprocessor saved:"
)

print(
    PREPROCESSOR_PATH
)


# =============================================================================
# 58. SAVE PROCESSED DATA
# =============================================================================

working_df.to_csv(

    PROCESSED_CSV_PATH,

    index=False
)


print(
    "\nProcessed CSV saved:"
)

print(
    PROCESSED_CSV_PATH
)


# =============================================================================
# 59. CREATE PREDICTION DATAFRAME
# =============================================================================

prediction_df = (
    X_test.copy()
)


prediction_df[
    "Actual_Class"
] = (
    y_test
    .map(
        reverse_mapping
    )
    .to_numpy()
)


prediction_df[
    "Random_Forest_Prediction"
] = (
    pd.Series(
        rf_predictions
    )
    .map(
        reverse_mapping
    )
    .to_numpy()
)


prediction_df[
    "Neural_Network_Prediction"
] = (
    pd.Series(
        nn_predictions
    )
    .map(
        reverse_mapping
    )
    .to_numpy()
)


# =============================================================================
# 60. ADD NN PROBABILITIES
# =============================================================================

prediction_df[
    "NN_Probability_Low"
] = nn_probabilities[
    :,
    0
]


prediction_df[
    "NN_Probability_Medium"
] = nn_probabilities[
    :,
    1
]


prediction_df[
    "NN_Probability_High"
] = nn_probabilities[
    :,
    2
]


# =============================================================================
# 61. SAVE PREDICTIONS
# =============================================================================

prediction_df.to_csv(

    TEST_PREDICTION_PATH,

    index=False
)


print(
    "\nTest predictions saved:"
)

print(
    TEST_PREDICTION_PATH
)


# =============================================================================
# 62. GET CATEGORY INFORMATION
# =============================================================================

mine_categories = []


if mine_column is not None:

    mine_categories = sorted(

        working_df[
            mine_column
        ]
        .astype(str)
        .unique()
        .tolist()
    )


company_categories = []


if company_column is not None:

    company_categories = sorted(

        working_df[
            company_column
        ]
        .astype(str)
        .unique()
        .tolist()
    )


# =============================================================================
# 63. JSON SERIALIZATION FUNCTION
# =============================================================================

def make_json_serializable(
    obj
):

    if isinstance(
        obj,
        dict
    ):

        return {

            str(key):
                make_json_serializable(
                    value
                )

            for key, value
            in obj.items()
        }


    if isinstance(
        obj,
        list
    ):

        return [

            make_json_serializable(
                value
            )

            for value in obj
        ]


    if isinstance(
        obj,
        tuple
    ):

        return [

            make_json_serializable(
                value
            )

            for value in obj
        ]


    if isinstance(
        obj,
        np.integer
    ):

        return int(
            obj
        )


    if isinstance(
        obj,
        np.floating
    ):

        return float(
            obj
        )


    if isinstance(
        obj,
        np.ndarray
    ):

        return obj.tolist()


    return obj


# =============================================================================
# 64. JSON METADATA
# =============================================================================

print("\n" + "=" * 80)
print("CREATING JSON METADATA")
print("=" * 80)


metadata = {

    "project": {

        "name": (
            "AI-Based Coal Mine Production "
            "Analysis and Performance Classification"
        ),

        "task":
            "Multiclass Classification",

        "version":
            "1.0"
    },


    "dataset": {

        "file_name":
            os.path.basename(
                CSV_PATH
            ),

        "dataset_path":
            CSV_PATH,

        "encoding":
            used_encoding,

        "rows":
            int(
                working_df.shape[0]
            ),

        "columns":
            int(
                working_df.shape[1]
            )
    },


    "detected_columns": {

        "mine":
            mine_column,

        "company":
            company_column,

        "production":
            production_column,

        "target":
            TARGET_COLUMN,

        "features":
            feature_columns
    },


    "production_statistics": {

        "minimum":
            float(
                working_df[
                    production_column
                ].min()
            ),

        "maximum":
            float(
                working_df[
                    production_column
                ].max()
            ),

        "mean":
            float(
                working_df[
                    production_column
                ].mean()
            ),

        "median":
            float(
                working_df[
                    production_column
                ].median()
            ),

        "standard_deviation":
            float(
                working_df[
                    production_column
                ].std()
            )
    },


    "classification": {

        "method":
            "Quantile-based classification",

        "classes": [
            "Low",
            "Medium",
            "High"
        ],

        "class_mapping":
            class_mapping,

        "low_medium_threshold":
            q33,

        "medium_high_threshold":
            q66
    },


    "random_forest": {

        "algorithm":
            "RandomForestClassifier",

        "accuracy":
            float(
                rf_accuracy
            ),

        "accuracy_percentage":
            float(
                rf_accuracy * 100
            ),

        "confusion_matrix":
            rf_cm.tolist(),

        "classification_report":
            make_json_serializable(
                rf_report
            )
    },


    "neural_network": {

        "framework":
            "TensorFlow/Keras",

        "input_dimension":
            int(
                input_dimension
            ),

        "hidden_layers": [
            128,
            64,
            32
        ],

        "output_classes":
            3,

        "test_loss":
            float(
                nn_loss
            ),

        "accuracy":
            float(
                nn_accuracy
            ),

        "accuracy_percentage":
            float(
                nn_accuracy * 100
            ),

        "confusion_matrix":
            nn_cm.tolist(),

        "classification_report":
            make_json_serializable(
                nn_report
            )
    },


    "categories": {

        "number_of_mines":
            len(
                mine_categories
            ),

        "number_of_companies":
            len(
                company_categories
            ),

        "companies":
            company_categories
    },


    "generated_files": {

        "pkl_model":
            os.path.basename(
                PKL_MODEL_PATH
            ),

        "h5_model":
            os.path.basename(
                H5_MODEL_PATH
            ),

        "yaml":
            os.path.basename(
                YAML_PATH
            ),

        "json":
            os.path.basename(
                JSON_PATH
            ),

        "preprocessor":
            os.path.basename(
                PREPROCESSOR_PATH
            ),

        "processed_csv":
            os.path.basename(
                PROCESSED_CSV_PATH
            ),

        "prediction_csv":
            os.path.basename(
                TEST_PREDICTION_PATH
            )
    }
}


# =============================================================================
# 65. SAVE JSON
# =============================================================================

with open(
    JSON_PATH,
    "w",
    encoding="utf-8"
) as file:

    json.dump(

        make_json_serializable(
            metadata
        ),

        file,

        indent=4,

        ensure_ascii=False
    )


print(
    "\nJSON saved:"
)

print(
    JSON_PATH
)


# =============================================================================
# 66. YAML CONFIGURATION
# =============================================================================

print("\n" + "=" * 80)
print("CREATING YAML CONFIGURATION")
print("=" * 80)


yaml_config = {

    "project": {

        "name": (
            "AI-Based Coal Mine Production "
            "Analysis and Performance Classification"
        ),

        "task":
            "Multiclass Classification",

        "version":
            "1.0"
    },


    "paths": {

        "project_directory":
            PROJECT_DIR,

        "dataset":
            CSV_PATH,

        "pkl_model":
            PKL_MODEL_PATH,

        "h5_model":
            H5_MODEL_PATH,

        "preprocessor":
            PREPROCESSOR_PATH,

        "metadata":
            JSON_PATH,

        "processed_data":
            PROCESSED_CSV_PATH,

        "predictions":
            TEST_PREDICTION_PATH
    },


    "columns": {

        "mine":
            mine_column,

        "company":
            company_column,

        "production":
            production_column,

        "features":
            feature_columns,

        "target":
            TARGET_COLUMN
    },


    "preprocessing": {

        "remove_empty_rows":
            True,

        "remove_empty_columns":
            True,

        "make_column_names_unique":
            True,

        "categorical_missing_value":
            "Unknown",

        "categorical_encoding":
            "OneHotEncoder",

        "handle_unknown":
            "ignore",

        "neural_network_scaler":
            "StandardScaler"
    },


    "target": {

        "method":
            "Quantile-based classification",

        "classes": {

            0:
                "Low",

            1:
                "Medium",

            2:
                "High"
        },

        "thresholds": {

            "low_medium":
                float(
                    q33
                ),

            "medium_high":
                float(
                    q66
                )
        }
    },


    "split": {

        "test_size":
            0.20,

        "random_state":
            42,

        "stratify":
            True
    },


    "random_forest": {

        "n_estimators":
            300,

        "max_depth":
            12,

        "min_samples_split":
            2,

        "min_samples_leaf":
            1,

        "class_weight":
            "balanced",

        "accuracy":
            float(
                rf_accuracy
            )
    },


    "neural_network": {

        "hidden_layers": [
            128,
            64,
            32
        ],

        "dropout": [
            0.30,
            0.20
        ],

        "activation":
            "relu",

        "output_activation":
            "softmax",

        "optimizer":
            "adam",

        "loss":
            "sparse_categorical_crossentropy",

        "epochs":
            150,

        "batch_size":
            16,

        "early_stopping_patience":
            15,

        "accuracy":
            float(
                nn_accuracy
            )
    }
}


# =============================================================================
# 67. SAVE YAML
# =============================================================================

with open(
    YAML_PATH,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(

        yaml_config,

        file,

        sort_keys=False,

        allow_unicode=True
    )


print(
    "\nYAML saved:"
)

print(
    YAML_PATH
)


# =============================================================================
# 68. VERIFY GENERATED FILES
# =============================================================================

print("\n" + "=" * 80)
print("VERIFYING GENERATED FILES")
print("=" * 80)


generated_files = [

    PKL_MODEL_PATH,

    H5_MODEL_PATH,

    YAML_PATH,

    JSON_PATH,

    PREPROCESSOR_PATH,

    PROCESSED_CSV_PATH,

    TEST_PREDICTION_PATH
]


for file_path in generated_files:

    if os.path.exists(
        file_path
    ):

        size = os.path.getsize(
            file_path
        )

        print(
            f"[OK] "
            f"{os.path.basename(file_path)} "
            f"- {size:,} bytes"
        )

    else:

        print(
            f"[MISSING] "
            f"{file_path}"
        )


# =============================================================================
# 69. FINAL RESULTS
# =============================================================================

print("\n" + "=" * 80)
print("FINAL RESULTS")
print("=" * 80)


print(
    f"""
Project:
AI-Based Coal Mine Production Analysis and Performance Classification


DATASET
-------
{CSV_PATH}


DETECTED COLUMNS
----------------
Mine:
{mine_column}

Company / Region:
{company_column}

Production:
{production_column}


FEATURES USED
-------------
{feature_columns}


IMPORTANT
---------
Production quantity is NOT used as an input feature.

It is used to create the Low / Medium / High
performance classes.

This prevents target leakage.


PRODUCTION CLASS THRESHOLDS
---------------------------
Low / Medium:
{q33:.6f}

Medium / High:
{q66:.6f}


MODEL RESULTS
-------------
Random Forest Accuracy:
{rf_accuracy * 100:.2f}%

Neural Network Accuracy:
{nn_accuracy * 100:.2f}%


GENERATED FILES
---------------
1. coal_production_model.pkl

2. coal_production_model.h5

3. coal_production_config.yaml

4. coal_production_metadata.json

5. coal_preprocessor.pkl

6. processed_coal_production_data.csv

7. test_predictions.csv


OUTPUT DIRECTORY
----------------
{PROJECT_DIR}
"""
)


print("=" * 80)
print("PROJECT COMPLETED SUCCESSFULLY")
print("=" * 80)
```
